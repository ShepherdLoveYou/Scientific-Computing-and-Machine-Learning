## MNIST数据库 / The MNIST Database

---

<img src="https://upload.wikimedia.org/wikipedia/commons/2/27/MnistExamples.png" style="width: 40%;"/>

MNIST（Modified National Institute of Standards and Technology）是一个大型的手写数字数据库。该数据库还广泛用于机器学习领域的训练和测试。MNIST 数据库包含 60,000 张训练图像和 10,000 张测试图像，均为28像素乘28像素的灰度图。

MNIST (Modified National Institute of Standards and Technology) is a large database of handwritten digits. The database is widely used for training and testing in the field of machine learning. The MNIST database contains 60,000 training images and 10,000 test images, all grayscale images of size 28 by 28 pixels.

<!-- bilingual -->

MNIST 数据集可以直接通过 urllib 下载（约 11 MB，首次运行时缓存到 `~/.cache/scml`）。本项目镜像未预装 TensorFlow / Keras / PyTorch，这样更轻量；学生如需在作业中训练模型，也是被要求手动用 NumPy 实现的。

MNIST can be fetched directly via urllib (about 11 MB, cached at `~/.cache/scml` on first run). This project image intentionally does not pre-install TensorFlow / Keras / PyTorch — the Ch15 assignment requires students to implement the network by hand in NumPy anyway.

<!-- bilingual -->

```python
import urllib.request, os
import numpy as np

# MNIST loader without torch/keras: downloads ~11 MB of data to ~/.cache/scml
# on first run, then loads from cache. Matches the numpy shapes that
# keras.datasets.mnist.load_data() used to return.
MNIST_URL = "https://storage.googleapis.com/tensorflow/tf-keras-datasets/mnist.npz"
cache = os.path.expanduser("~/.cache/scml/mnist.npz")
os.makedirs(os.path.dirname(cache), exist_ok=True)
if not os.path.exists(cache):
    urllib.request.urlretrieve(MNIST_URL, cache)
with np.load(cache, allow_pickle=True) as f:
    X_train, y_train = f["x_train"], f["y_train"]
    X_test,  y_test  = f["x_test"],  f["y_test"]
print(X_train.shape, y_train.shape, X_test.shape, y_test.shape)
```

也可以使用Numpy直接读取存放在本地的mnist.npz文件。

The local file `mnist.npz` can also be read directly using NumPy.

<!-- bilingual -->

```python
import numpy as np
from pathlib import Path

# Alternative path: load from a local mnist.npz file if one is present
# (e.g. downloaded once and shipped alongside the notebook). Falls back to
# the keras-loaded data above when the local file is absent.
_local = Path('mnist.npz')
if _local.exists():
    f = np.load(_local)
    X_train, y_train = f['x_train'], f['y_train']
    X_test, y_test = f['x_test'], f['y_test']
    f.close()
else:
    print('mnist.npz not found locally; using data loaded above via keras.datasets.mnist')
```

### 基于机器学习的手写数字识别 / Handwritten Digit Recognition Based on Machine Learning

---

手写字识别是一个数据分类问题。MNIST数据集提供的训练样本已经进行了人工标记，我们可以选择一种分类器进行监督学习，例如scikl-learn库中提供的线性分类器、支持向量机、K-最近邻、决策树、随机森林等，可以得到不错的模型准确率（>98%）。

Handwritten digit recognition is a data classification problem. The training samples provided by the MNIST dataset have been manually labeled, so we can choose a classifier for supervised learning — for example, the linear classifiers, support vector machines, k-nearest neighbors, decision trees, and random forests provided by the scikit-learn library — and obtain a decent model accuracy (>98%).

神经网络作为一种更为强大的工具，可以实现更高的分类准确率。

As a more powerful tool, neural networks can achieve even higher classification accuracy.
