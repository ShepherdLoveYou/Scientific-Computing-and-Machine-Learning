Seaborn是Python中的一个高级可视化库，是对Matplotlib进行二次封装而成。Seaborn的很多图表接口和参数设置与Matplotlib很是接近。相比Matplotlib而言，Seaborn的几个鲜明特点如下：

Seaborn is a high-level visualization library in Python built on top of Matplotlib. Many of Seaborn's chart interfaces and parameter settings closely resemble those of Matplotlib. Compared with Matplotlib, Seaborn has the following distinctive features:

* 绘图接口更为集成，可通过少量参数设置实现大量封装绘图

* More integrated plotting interfaces: large amounts of plotting can be achieved with only a few parameter settings

* 多数图表具有统计学含义，例如分布、关系、统计、回归等

* Most charts have statistical meaning, such as distribution, relationship, statistics, regression, etc.

* 对Pandas和Numpy数据类型支持非常友好

* Very friendly support for Pandas and Numpy data types

* 风格设置更为多样，例如风格、绘图环境和颜色配置等

* More varied style configuration, such as themes, plotting context, and color palettes

# Ipyvolume
---

IPyvolume is a Python library to visualize 3d volumes and glyphs (e.g. 3d scatter plots), in the Jupyter notebook, with minimal configuration and effort. It is currently pre-1.0, so use at own risk. IPyvolume’s volshow is to 3d arrays what matplotlib’s imshow is to 2d arrays.

IPyvolume is a Python library to visualize 3d volumes and glyphs (e.g. 3d scatter plots), in the Jupyter notebook, with minimal configuration and effort. It is currently pre-1.0, so use at own risk. IPyvolume's volshow is to 3d arrays what matplotlib's imshow is to 2d arrays.

## 参考资料 / References

[Ipyvolume 手册](https://ipyvolume.readthedocs.io/en/latest/index.html)

[Ipyvolume manual](https://ipyvolume.readthedocs.io/en/latest/index.html)

[Ipyvolume 示例](https://ipyvolume.readthedocs.io/en/latest/examples.html)

[Ipyvolume examples](https://ipyvolume.readthedocs.io/en/latest/examples.html)

<!-- bilingual -->

```python
import numpy as np
import ipyvolume as ipv
V = np.zeros((128,128,128)) # our 3d array
# outer box
V[30:-30,30:-30,30:-30] = 0.75
V[35:-35,35:-35,35:-35] = 0.0
# inner box
V[50:-50,50:-50,50:-50] = 0.25
V[55:-55,55:-55,55:-55] = 0.0

ipv.figure()
ipv.volshow(V, level=[0.25, 0.75], opacity=0.03, level_width=0.1, data_min=0, data_max=1)
ipv.view(-30, 40)
ipv.show()
```

# Bqplot
---

Bqplot是用于Jupyter的交互式2D绘图库，其中绘图的每个属性都是一个交互式小部件，只需几行Python代码就可以创建丰富的可视化效果。Bqplot构建在widgets框架之上，它利用widget基础提供第一个在Python和JavaScript代码之间通信的绘图库。Bqplot的可视化基于D3.js和SVG，支持快速交互和漂亮的动画。

Bqplot is an interactive 2D plotting library for Jupyter, in which every property of a plot is an interactive widget, so rich visualizations can be created with only a few lines of Python code. Bqplot is built on top of the widgets framework, which uses the widget infrastructure to provide the first plotting library for communication between Python and JavaScript code. Bqplot's visualizations are based on D3.js and SVG, and support fast interactions and beautiful animations.

## 参考资料 / References

[bqplot 手册](https://bqplot.readthedocs.io/en/latest)

[bqplot manual](https://bqplot.readthedocs.io/en/latest)

[bqplot 示例](https://github.com/bqplot/bqplot/tree/master/examples)

[bqplot examples](https://github.com/bqplot/bqplot/tree/master/examples)

<!-- bilingual -->

```python
import numpy as np
from bqplot import pyplot as plt

size = 100
np.random.seed(0)
x_data = np.arange(size)
y_data = np.cumsum(np.random.randn(size))

figure = plt.figure(title='Bqplot Plot')
scatter = plt.scatter(x_data, y_data)
plt.show()
```

```python
scatter.y = np.cumsum(np.random.randn(size))  # 更新数据
```

```python
scatter.colors = ['Green']  # 更新颜色
```

### 动画 / Animation

<!-- bilingual -->

```python
import time

for i in range(100):  # Update Chart
    scatter.y = np.insert(scatter.y[:-1], 0 , scatter.y[-1])
    time.sleep(0.1)
```

### 交互 / Interaction

<!-- bilingual -->

```python
import numpy as np
from bqplot import pyplot as plt
from ipywidgets import interactive, FloatSlider, jslink

x = np.linspace(-5, 5, 100)
fig = plt.figure(title='Bqplot Plot')
line = plt.plot(x, np.sin(x))

plt.set_lim(-2, 2, 'y')

plt.show()

def f(a, b):
    line.y = a * np.sin(b*x)

interact_plot = interactive(f, a=(1.0, 2.0), b=(1.0, 5.0))
interact_plot
```
