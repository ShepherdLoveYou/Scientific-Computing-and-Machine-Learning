# 随机过程和等概率原理 / Stochastic Processes and the Equal-Probability Principle

---

## 游戏1 / Game 1

一个城镇1000个人共同玩一个游戏。市民初始资金为20元。每轮游戏中，随机抽取一个市民A，则让其把自己的1元交给同样随机抽取的B。市民的资金量可以为负数。

A town of 1,000 people plays a game together. Every resident starts with 20 dollars. In each round of the game, a resident A is randomly selected and told to give 1 dollar to a resident B, also randomly selected. A resident's balance can be negative.

在这样的规则下，每个市民有相同的概率失去1元，也有相同的概率得到1元。

Under these rules, every resident has an equal probability of losing 1 dollar and an equal probability of gaining 1 dollar.

当重复多论以后，我们再来查看每个人持有资金的分布。

After many rounds, we then examine the distribution of each person's balance.

<!-- bilingual -->

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

num_of_people = 1000
money_list = np.zeros(num_of_people)+10
for i in range(int(2e6)):
    n = int(np.random.rand()*num_of_people)
    m = int(np.random.rand()*num_of_people)
    money_list[n] -= 1
    money_list[m] += 1

plt.subplot(1,2,1)
plt.xlabel('Sorted List')
plt.ylabel('Money')
plt.plot(np.sort(money_list)) # 依照资金量从低到高排序
plt.subplot(1,2,2)
plt.xlabel('Money')
plt.ylabel('Number of People')
plt.hist(money_list, rwidth=5) # 持有特定资金量的人数统计
plt.tight_layout()
```

上图为随机过程中常见的**正态分布**。

The figure above shows a **normal distribution**, which is common in stochastic processes.

## 游戏2 / Game 2

重复上述的游戏，并增加一个规则。如果市民持有的资金等于0元，即使被抽中不会输掉赌注。

Repeat the game above, but add one rule: if a resident's balance equals 0, being selected does not make them lose money.

当重复多论以后，我们再来查看每个人持有资金的分布。

After many rounds, we then examine the distribution of each person's balance.

<!-- bilingual -->

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

num_of_people = 1000
money_list = np.ones(num_of_people)*10
for i in range(int(2e6)):
    n = int(np.random.rand()*num_of_people)
    if money_list[n] > 0:
        m = int(np.random.rand()*num_of_people)
        money_list[n] -= 1
        money_list[m] += 1

plt.subplot(1,2,1)
plt.xlabel('Sorted List')
plt.ylabel('Money')
plt.plot(np.sort(money_list)) # 依照资金量从低到高排序
plt.subplot(1,2,2)
plt.xlabel('Money')
plt.ylabel('Number of People')
plt.hist(money_list, rwidth=5) # 持有特定资金量的人数统计
plt.tight_layout()
```

按照资金量划分阶层，得到的人数统计为玻尔兹曼分布。
$$F({\rm class}) \propto e^{-\frac{Money}{\alpha}}$$
其中系数$\alpha$等于平均资金量。

Grouping residents into strata by balance gives a head-count distribution that follows the Boltzmann distribution.
$$F({\rm class}) \propto e^{-\frac{Money}{\alpha}}$$
where the coefficient $\alpha$ equals the average balance.

## 物理系统中的等概率原理 / The Equal-Probability Principle in Physical Systems

---

将市民类比为一个物理系统的微观状态，资金类比为这个微观状态具有的能量，我们可以得到统计物理学的重要的基本假定，等概率原理。等概率原理指当系统处于平衡时，如果除能量一定、体积一定和粒子数一定外，没有任何其他的限制，则发现系统处在各微观状态的概率都是相同的。

If we view each resident as analogous to a microstate of a physical system and the balance as analogous to the energy of that microstate, we obtain an important fundamental postulate of statistical physics: the equal-probability principle. The equal-probability principle states that for a system in equilibrium, if there are no restrictions other than fixed energy, fixed volume, and fixed particle number, then the probability of finding the system in any given microstate is the same.

等概率原理本质上等价于这样的论断：孤立系在平衡态下具有最大熵值，即熵极大原理。因此，我们可以认为等概率原理与热力学第二定律具有等价性。

The equal-probability principle is essentially equivalent to the statement that an isolated system at equilibrium attains maximum entropy, i.e. the principle of maximum entropy. Therefore, the equal-probability principle can be considered equivalent to the second law of thermodynamics.

### 麦克斯韦分布 / Maxwell Distribution

数学上推导麦克斯韦分布律需要使用玻尔兹曼分布。在统计力学中，玻尔兹曼分布是系统中的粒子在平衡态时关于能量的统计分布：
$$F({\rm state}) \propto e^{-\frac{E}{kT}}$$

Mathematically deriving the Maxwell distribution requires use of the Boltzmann distribution. In statistical mechanics, the Boltzmann distribution is the equilibrium statistical distribution of particles in a system as a function of energy:
$$F({\rm state}) \propto e^{-\frac{E}{kT}}$$

其中$E$为能量。在不引入其它假设（*最大熵假说/等概率假说/各态历经假说*）的前提下，玻尔兹曼分布*无法*从经典力学导出。

where $E$ is the energy. Without introducing additional assumptions (*maximum-entropy hypothesis / equal-probability hypothesis / ergodic hypothesis*), the Boltzmann distribution *cannot* be derived from classical mechanics.
