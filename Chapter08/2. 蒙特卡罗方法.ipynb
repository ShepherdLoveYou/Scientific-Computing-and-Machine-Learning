# 蒙特卡罗算法 / Monte Carlo Algorithms

---

[蒙特卡罗方法](https://zh.wikipedia.org/wiki/蒙地卡羅方法)（Monte Carlo method），也称统计模拟方法，是一种以概率统计理论为指导的数值计算方法。它使用随机数（或更常见的伪随机数）来解决科学和工程中的很多计算问题。

The [Monte Carlo method](https://en.wikipedia.org/wiki/Monte_Carlo_method), also known as statistical simulation, is a numerical computation approach guided by probability and statistics theory. It uses random numbers (or more commonly pseudorandom numbers) to solve many computational problems in science and engineering.

通常蒙特卡罗方法可以粗略地分成两类：

Monte Carlo methods can be roughly divided into two categories:

1. 一类是所求解的问题本身具有内在的随机性，借助计算机的运算能力可以直接模拟这种随机的过程。例如模拟核裂变过程。
2. 另一种类型是所求解问题可以转化为某种随机分布的特征数，比如随机事件出现的概率，或者随机变量的期望值。通过随机抽样的方法，以随机事件出现的频率估计其概率，或者以抽样的数字特征估算随机变量的数字特征，并将其作为问题的解。例如求解复杂的多维积分问题。

1. One category is problems that have inherent randomness; the computer's computational power can be used to directly simulate such random processes. For example, simulating nuclear fission.
2. The other category is problems that can be transformed into a characteristic quantity of some random distribution, such as the probability of a random event or the expected value of a random variable. Random sampling is used to estimate the probability of a random event from its frequency, or to estimate characteristic quantities of a random variable from sampled numerical characteristics, which are then taken as the solution to the problem. For example, solving complex high-dimensional integration problems.

**示例** 使用蒙特卡罗方法估算$\pi$值

**Example** Use the Monte Carlo method to estimate the value of $\pi$.

<!-- bilingual -->

```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(4, 4))
x = np.linspace(0,1,100)
y = np.sqrt(1 - x**2)
ax.plot(x, y, c='b')
ax.fill_between(x, 0, y, alpha=0.5)
ax.fill_between(x, y, 1, alpha=0.5)
```

```python
from ipywidgets import interactive

np.random.seed(0)

def pi(n):
    N = 10**n
    x = np.random.random(N)
    y = np.random.random(N)
    dist = x**2 + y**2
    pi = np.sum(dist < 1) / N * 4
    X = np.linspace(0,1,100)
    Y = np.sqrt(1 - X**2)
    fig, ax = plt.subplots(figsize=(4, 4))
    ax.plot(X, Y, c='b')
    ax.scatter(x, y, s=1, alpha=0.5)
    plt.show()
    print('Number of Data Point  %d' %(N))
    print('Estimate Value of Pi  %f' %(pi))

interact_plot = interactive(pi, n=(1, 7))
interact_plot
```

## 伪随机数生成器（PRNG） / Pseudorandom Number Generators (PRNG)

---

经典计算机的数据都是由确定性算法生成的，因此随机数生成算法只能得到伪随机数序列。虽然伪随机数并不真正的随机，但具有类似于随机数的统计特征，如均匀性、独立性等。真正的随机数必须使用专门的设备，比如热噪信号、用户按键盘的位置与速度、移动设备加速度传感器等，或者使用量子计算机。

All data on a classical computer are generated by deterministic algorithms, so random-number generation algorithms can only produce pseudorandom number sequences. Although pseudorandom numbers are not truly random, they have statistical characteristics similar to those of random numbers, such as uniformity and independence. True random numbers must be obtained from specialized devices, such as thermal noise signals, the position and speed at which a user presses keys on a keyboard, or accelerometers in mobile devices, or must be produced by quantum computers.

密码学中使用伪随机数要小心，其可计算性是一个可以攻击的地方。统计学、蒙特卡罗方法上使用的伪随机数也必须挑选周期极长、随机性够高的随机函数，以确保计算结果有足够的随机性。

Using pseudorandom numbers in cryptography requires caution because their computability is an avenue for attack. Pseudorandom numbers used in statistics and Monte Carlo methods must also be chosen to have extremely long periods and high randomness to ensure that the results have sufficient randomness.

### 线性同余发生器（LCG） / Linear Congruential Generator (LCG)

[线性同余发生器](https://zh.wikipedia.org/wiki/線性同餘方法)是一种能产生具有不连续计算的**均匀分布**伪随机序列的分段线性方程的算法。它的递推公式是：
$$N_{{j+1}}\equiv (A\times N_{j}+B){\pmod  {M}}$$
其中参数为乘数$A$、增量$B$和模数$M$。

A [linear congruential generator](https://en.wikipedia.org/wiki/Linear_congruential_generator) is an algorithm that uses a piecewise linear equation with discontinuous calculations to produce a pseudorandom sequence with a **uniform distribution**. Its recurrence relation is:
$$N_{{j+1}}\equiv (A\times N_{j}+B){\pmod  {M}}$$
where the parameters are the multiplier $A$, the increment $B$, and the modulus $M$.

Hull-Dobell定理：当且仅当满足以下条件时，LCG序列的周期最大（最大周期为M）：

The Hull-Dobell theorem states that the period of an LCG sequence is maximized (with maximum period equal to $M$) if and only if:

1. $B$和$M$互质
2. $A-1$可以被$M$的所有质因数整除
3. 如果$M$是4的倍数，则$A-1$也需要是4的倍数

1. $B$ and $M$ are coprime.
2. $A - 1$ is divisible by every prime factor of $M$.
3. If $M$ is a multiple of 4, then $A - 1$ must also be a multiple of 4.

序列的初始值$N_0$被称为*种子*。LCG算法通常被设计为返回$z/m$，即[0, 1]区间的浮点数。

The initial value $N_0$ of the sequence is called the *seed*. LCG algorithms are usually designed to return $z/m$, a floating-point number in the interval [0, 1].

<!-- bilingual -->

```python
def lcg(m=2**32, a=1103515245, b=12345):
    lcg.current = (a*lcg.current + b) % m
    return lcg.current/m

# setting the seed
lcg.current = 1

[lcg() for i in range(10)]
```

### 逆变换采样 / Inverse Transform Sampling

[逆变换采样](https://zh.m.wikipedia.org/zh-hans/逆变换采样)是伪随机数采样的一种基本方法。在已知**任意概率分布**的累积分布函数时，可用于从该分布中生成随机样本。

[Inverse transform sampling](https://en.wikipedia.org/wiki/Inverse_transform_sampling) is a basic method for pseudorandom sampling. When the cumulative distribution function of an **arbitrary probability distribution** is known, it can be used to generate random samples from that distribution.

<!-- bilingual -->

```python
def expon_pdf(x, lmabd=1):
    """指数分布的概率密度函数"""
    return lmabd*np.exp(-lmabd*x)
```

```python
def expon_cdf(x, lambd=1):
    """指数分布的累积分布函数"""
    return 1 - np.exp(-lambd*x)
```

```python
def expon_icdf(p, lambd=1):
    """指数分布的累积分布函数逆函数，即分位函数"""
    return -np.log(1-p)/lambd
```

## 蒙特卡罗积分 / Monte Carlo Integration

---

假设我们需要进行如下积分：

Suppose we need to compute the following integral:

$$I = \int_a^b f(x) dx$$

积分体积为$V$。蒙特卡罗积分通过计算落在$f(x)$中随机分布点的比例与$V$的乘积来估计积分值的近似值。

The volume of integration is $V$. Monte Carlo integration estimates an approximation of the integral by computing the proportion of randomly distributed points that fall under $f(x)$, multiplied by $V$.

从统计学角度出发，我们将上式改写为：

From a statistical perspective, we can rewrite the above as:

$$I = \int_X \frac{f(x)}{p(x)}p(x)dx = \int_X h(x)p(x)dx = E[h(X)]$$

计算积分等价于找到$h(X)=\frac{f(x)}{p(x)}$的数学期望值。对于一维单变量问题，$p(x)=1/(b-a)$。

Computing the integral is equivalent to finding the expectation of $h(X) = \frac{f(x)}{p(x)}$. For one-dimensional univariate problems, $p(x) = 1/(b - a)$.

我们可以使用$n$次随机采样，对该期望值进行估计，

We can estimate this expectation by using $n$ random samples,

$$E[h(X)] \approx \bar{h_n} = \frac{1}{n} \sum_{i=1}^n h(x_i) $$

我们可以对蒙特卡罗方法的方差进行估计，

We can also estimate the variance of the Monte Carlo method,

$$v_n = \frac{1}{n^2} \sum_{o=1}^n (h(x_i) - \bar{h_n})^2$$

$$v_n = \frac{1}{n^2} \sum_{i=1}^n (h(x_i) - \bar{h_n})^2$$

根据[中心极限定理](https://zh.wikipedia.org/wiki/中心极限定理)，

According to the [central limit theorem](https://en.wikipedia.org/wiki/Central_limit_theorem),

$$\frac{\bar{h_n} - E[h(X)]}{\sqrt{v_n}} \sim \mathcal{N}(0, 1)$$

可知蒙特卡罗积分的收敛速度是$\mathcal{0}(n^{1/2})$，且与维度无关。因此，相对于数值积分（收敛速度是$\mathcal{0}(n^{d})$），蒙特卡罗积分在处理高维问题时更为有效。

So the convergence rate of Monte Carlo integration is $\mathcal{O}(n^{1/2})$ and is independent of dimension. Therefore, relative to numerical integration (whose convergence rate is $\mathcal{O}(n^d)$), Monte Carlo integration is more effective when handling high-dimensional problems.

```python
# Monte Carlo approximation
for n in 10**np.array([1,2,3,4,5,6]):
    pts = np.random.uniform(0, 1, (n, 2))
    pts[:, 1] *= np.e
    count = np.sum(pts[:, 1] < np.exp(pts[:, 0]))
    volume = np.e * 1 # volume of region
    sol = (volume * count)/n
    print('%10d %.6f' % (n, sol))
```

**Law of Large Numbers** The greater the number of samples, the more likely their arithmetic mean is to approach the expected value. ### 方差估计 / Variance Estimation

蒙特卡罗积分的结果是否收敛对于结果的准确性非常重要。因此，我们需要对方差进行估计。一个获取结果置信区间的简单方法是重复进行多次蒙特卡罗模拟。

Whether the result of Monte Carlo integration has converged is very important for accuracy. Therefore, we need to estimate the variance. A simple way to obtain a confidence interval of the result is to repeat the Monte Carlo simulation multiple times.

<!-- bilingual -->

```python
n = 10**7
sol = []

for i in range(10):
    pts = np.random.uniform(0, 1, (n, 2))
    pts[:, 1] *= np.e
    count = np.sum(pts[:, 1] < np.exp(pts[:, 0]))
    volume = np.e * 1 
    sol.append((volume * count)/n)

np.percentile(sol, [2.5, 50, 97.5])
```

### 方差缩减 / Variance Reduction

蒙特卡罗积分的期望误差与$\sqrt n$成反比，精度可以通过增加模拟的次数，但是速度较慢。提高计算精度更有效的途径是减小方差。常见的方差缩减技术有分层抽样法、重要性抽样法、条件期望法、对偶变量法、控制变量法、准随机数等。

The expected error of Monte Carlo integration is inversely proportional to $\sqrt{n}$. Accuracy can be improved by increasing the number of simulations, but this is slow. A more effective way to improve accuracy is to reduce the variance. Common variance-reduction techniques include stratified sampling, importance sampling, conditional expectation, antithetic variates, control variates, quasi-random numbers, etc.

**Importance Sampling**

使用常规蒙特卡罗方法计算标准正态分布的尾部几率$P(X > 5)$是非常困难的。因为$10^7$个随机采样中，只有平均3个采样大于5。

It is very difficult to use a standard Monte Carlo method to compute the tail probability $P(X > 5)$ of the standard normal distribution. Out of $10^7$ random samples, on average only 3 are greater than 5.

<!-- bilingual -->

```python
n = 10**6
y = ss.norm().rvs(n)
h_mc = 1.0/n * np.sum(y > 5)
print(h_mc)
```

我们可以引入另外一个分布函数$g(x)$做为重要性函数，使有限的采样更合理地分配在每个区间。蒙特卡罗积分地表达式改写为：

We can introduce another distribution function $g(x)$ as an importance function, so that a finite set of samples is distributed more reasonably across each interval. The Monte Carlo integration expression is rewritten as:

$$E_f[h(x)] \ = \  \int_X h(x) \frac{f(x)}{g(x)} g(x) dx \ = \ E_g\left[ \frac{h(X) f(X)}{g(X)} \right]$$

新的期望值为

The new expectation becomes

$$E[h(X)] \approx \bar{h_n} = \frac{1}{n} \sum_{i=1}^n \frac{f(x_i)}{g(x_i)} h(x_i) $$

其中$f(x_i)/g(x_i)$标识出了样本$h(x_i)$在估计$\bar{h_n}$中重要性。

where $f(x_i)/g(x_i)$ indicates the importance of sample $h(x_i)$ in estimating $\bar{h_n}$.

<!-- bilingual -->

```python
n = 10**4
y = ss.expon(loc=5).rvs(n)
h_is = 1.0/n * np.sum(ss.norm().pdf(y)/ss.expon(loc=5).pdf(y))
print(h_is)
```

### 无穷积分 / Infinite Integrals

蒙特卡罗积分可以计算与分布函数相关的无穷积分，例如下式：

Monte Carlo integration can compute infinite integrals related to distribution functions, such as the following:

$$\int_{-\infty}^{\infty} x^2 \frac{1}{\sqrt{2\pi}} \exp(\frac{-x^2}{2}) dx$$

我们取$f(x) = x^2$，则剩余部分$p(x)= \frac{1}{\sqrt{2\pi}} \exp(\frac{-x^2}{2})$是标准正态分布的密度函数。上述积分等价于计算函数$f(x)$在标准正态分布下的均值。

Taking $f(x) = x^2$, the remaining part $p(x) = \frac{1}{\sqrt{2\pi}} \exp(\frac{-x^2}{2})$ is the density function of the standard normal distribution. The above integral is equivalent to computing the mean of the function $f(x)$ under the standard normal distribution.

因此，我们可以按照标准正态分布在区间$(-\infty, \infty)$进行抽样，然后计算函数均值作为上述无穷积分的估计值。

Therefore, we can sample according to the standard normal distribution on the interval $(-\infty, \infty)$ and then compute the function mean as the estimate of the above infinite integral.

<!-- bilingual -->

```python
def f(x):
    return x**2

n = 10**5
pts = np.random.randn(n)
np.mean(f(pts))
```

### 多重积分 / Multiple Integrals

当增加积分变量时，数值方法计算多重积分的计算量（正比于$n^d$）快速增长。

As the number of integration variables increases, the computational cost of numerical methods for multiple integrals (proportional to $n^d$) grows rapidly.

<!-- bilingual -->

```python
def f(*args):
    return  np.exp(-np.sum(np.array(args)**2))

from scipy import integrate

%time print(integrate.nquad(f, [(0,1)] * 1))
%time print(integrate.nquad(f, [(0,1)] * 2))
%time print(integrate.nquad(f, [(0,1)] * 3))
%time print(integrate.nquad(f, [(0,1)] * 4))
```

蒙特卡罗积分的精度较低，但是在维度上的拓展性非常好。

Monte Carlo integration has lower precision, but it scales very well with dimensionality.

<!-- bilingual -->

```python
def mc(n, d):
    pts = np.random.uniform(0, 1, (n, d))
    sum = 0
    for i in pts:
        sum += f(i)
    return sum/len(pts)

n = 10**5

%time print(mc(n, 1))
%time print(mc(n, 2))
%time print(mc(n, 3))
%time print(mc(n, 4))
```

## Vegas 高维函数蒙特卡洛积分库 / Vegas: High-Dimensional Monte Carlo Integration Library

---

Python的vegas库使用vegas算法计算任意多维积分的蒙特卡洛估计。

Python's `vegas` library uses the VEGAS algorithm to compute Monte Carlo estimates of arbitrary multi-dimensional integrals.

该算法有两个步骤：
1. 一个自动转换被应用于积分变量，从而使得积分趋于平坦。
2. 用转换后的变量对积分进行蒙特卡洛估计。

The algorithm has two steps:
1. An automatic transformation is applied to the integration variables so that the integrand becomes flatter.
2. The integral is estimated by Monte Carlo using the transformed variables.

扁平化积分使积分更容易，并改善估计。应用于积分变量的变换会在迭代优化，一次迭代中收集的关于积分的信息被用来改进下一次迭代中使用的变换。

Flattening the integrand makes the integral easier and improves estimates. The transformation applied to the integration variables is iteratively optimized; information about the integral gathered in one iteration is used to improve the transformation used in the next.

## 蒙特卡罗数值优化 / Monte Carlo Numerical Optimization

### 模拟退火算法 / Simulated Annealing

[模拟退火](https://zh.wikipedia.org/wiki/模拟退火)算法来源于材料科学中的退火过程。当固体温度增加时，内部原子运动速度增加，固体内原子构型发生改变；当其缓慢冷却时，内部原子构型趋向有序，且内能（原子间相互作用能）减小。可以将系统内能写为原子坐标的函数$E(X_1, X_2,...,X_n)$，在退火后新的平衡位置上，$E$为局部或全局极小值。

The [simulated annealing](https://en.wikipedia.org/wiki/Simulated_annealing) algorithm is inspired by the annealing process in materials science. As the temperature of a solid increases, the speed of internal atomic motion increases and the internal atomic configuration changes; as it cools down slowly, the internal atomic configuration tends to become ordered and the internal energy (the interaction energy between atoms) decreases. The internal energy of the system can be written as a function of atomic coordinates $E(X_1, X_2, \ldots, X_n)$, and after annealing, at the new equilibrium, $E$ is at a local or global minimum.

在退火模拟中，每一步会随机生成一个新的原子构型。根据[Metropolis抽样](https://zh.m.wikipedia.org/zh-hans/梅特罗波利斯－黑斯廷斯算法)准则，新构型的接受概率为：
$$P(acc)=min(1,e^{-{\frac {\Delta U}{k_{B}T}}})$$
其中$\Delta U$为内能改变量，$k_{B}$为玻尔兹曼常数，$T$为温度。$k_{B}T$正比于原子的平均动能。

In simulation of annealing, each step randomly generates a new atomic configuration. According to [Metropolis sampling](https://en.wikipedia.org/wiki/Metropolis%E2%80%93Hastings_algorithm), the acceptance probability of the new configuration is:
$$P(acc)=min(1,e^{-{\frac {\Delta U}{k_{B}T}}})$$
where $\Delta U$ is the change in internal energy, $k_B$ is the Boltzmann constant, and $T$ is the temperature. $k_B T$ is proportional to the mean kinetic energy of the atoms.

优化问题的模拟退火算法和材料热退火过程有诸多相似之处。

The simulated annealing algorithm for optimization problems and the thermal annealing process of materials have many similarities.

| 模拟退火 | 物理退火 | 
|---|---|
| 变量 | 粒子状态 |
| 目标函数 | 能量 |
| 最优解 | 能量最低态 |
| 设定动能（控制参数） | 材料熔解 |
| Metropolis采样 | 等温过程 |
| 控制参数下降 | 材料冷却 |

| Simulated Annealing | Physical Annealing |
|---|---|
| Variable | Particle state |
| Objective function | Energy |
| Optimal solution | Lowest-energy state |
| Setting kinetic energy (control parameter) | Material melting |
| Metropolis sampling | Isothermal process |
| Decreasing control parameter | Material cooling |

模拟退火算法也常用于机器学习，特别是强化学习的算法中。一般情况下，针对得到的样本数据集创建相对模糊的模型，通过蒙特卡洛方法对于模型中的参数进行选取，使之于原始数据的残差尽可能的小。从而达到创建模型拟合样本的目的。

Simulated annealing is also commonly used in machine learning, especially in reinforcement learning algorithms. In general, for a given sample dataset, a relatively blurry model is created, and Monte Carlo methods are used to choose parameters of the model so that the residual against the original data is as small as possible. The goal is to create a model that fits the samples.

### 函数优化 / Function Optimization

使用蒙特卡罗模拟退火算法进行函数优化是按照以下步骤进行的：

Using the Monte Carlo simulated annealing algorithm to optimize a function proceeds as follows:

1. 使用随机数生成器在变量空间中产生一个随机的位置坐标。
2. 对此位置做无规则的改变，产生一个新的位置坐标。
3. 计算新的位置坐标处的函数值。
4. 比较新的位置坐标与改变前的位置坐标的函数值变化，判断是否接受该坐标。
    * 若新的位置坐标函数值低于原位置坐标的函数值，则接受新的坐标，使用这个坐标重复再做下一次迭代。
    * 若新的位置坐标函数值高于原位置坐标的函数值，则计算玻尔兹曼因子，并产生一个随机数。
         * 若这个随机数大于所计算出的玻尔兹曼因子，则放弃这个坐标，重新计算。
         * 若这个随机数小于所计算出的玻尔兹曼因子，则接受这个坐标，使用这个坐标重复再做下一次迭代。
5. 如此进行迭代计算，直至最后搜索出低于所给函数值条件的位置坐标结束。

1. Use a random number generator to produce a random position in the variable space.
2. Make a random change to this position to produce a new position.
3. Compute the function value at the new position.
4. Compare the function value at the new position with that at the previous position to decide whether to accept the new position.
    * If the function value at the new position is lower than at the previous position, accept the new position and use it in the next iteration.
    * If the function value at the new position is higher than at the previous position, compute the Boltzmann factor and generate a random number.
         * If the random number is greater than the computed Boltzmann factor, discard the new position and recompute.
         * If the random number is less than the computed Boltzmann factor, accept the new position and use it in the next iteration.
5. Iterate the computation in this way, until the search yields a position with a function value lower than the given condition.

<!-- bilingual -->

```python
import numpy as np

x = np.linspace(-1, 1, 200)
y = np.linspace(-1, 1, 200)
z = np.linspace(-1, 1, 200)
Y, Z, X = np.meshgrid(x, y, z)

def gaussian(x0, y0, z0):
    return np.exp(-(X - x0)**2 - (Y - y0)**2 - (Z - z0)**2)

# 局部极小值在(-0.5, -0.5, -0.5)附近，全局极小值在(0.5, 0.5, 0.5)附近
data = gaussian(0.1, -0.1, -0.1) - gaussian(-0.5, -0.5, -0.5) - gaussian(0.5, 0.5, 0.5) * 2

np.unravel_index(np.argmin(data), data.shape), np.min(data)
```

```python
def annealing(pos, step, KbT):
    for i in range(step):
        value = data[tuple(pos)]
        new_pos = pos + np.round(np.random.random(3)) * 2 - 1
        new_pos = np.int32(new_pos % 200)  # pbc
        new_value = data[tuple(new_pos)]
        if new_value < value:
            pos = new_pos
            value = new_value
        else:
            delta = new_value - value
            p = np.exp(-delta/KbT)
            if p > np.random.random():
                pos = new_pos
                value = new_value
    return pos, value

step = 10**5
KbT = 0.01

#np.random.seed(0)

pos1 = [50, 50, 50]  # 初始位置在局部极小附近
print(annealing(pos1, step, KbT))

pos1 = [150, 150, 150]  # 初始位置在全局极小附近
print(annealing(pos1, step, KbT))
```

可以看到，在温度/动能较低时，蒙特卡罗模拟不能保证系统在有限的时间内收敛到最低点。增加模拟时间或者增加温度，可以增大系统从局部势阱约束中逃逸的几率。

It can be observed that, at low temperature / low kinetic energy, Monte Carlo simulations cannot guarantee that the system will converge to the lowest point within finite time. Increasing simulation time or increasing temperature can increase the probability that the system escapes from local potential wells.

<!-- bilingual -->

```python
'''变温'''
KbT = 0.1
pos = np.int32(np.random.random(3)*200)

for i in range(5):
    pos, value = annealing(pos, step, KbT / 10**i)
    print(KbT, pos, value)
```

1. 相对于数值算法只能收敛到局部最优解，模拟退火算法所得解依概率收敛到全局最优解。
2. 模拟退火算法对变量数目不敏感，可以很轻松处理高维问题。

1. Unlike numerical algorithms that can only converge to local optima, the solution from simulated annealing converges to the global optimum with probability.
2. Simulated annealing is insensitive to the number of variables and can handle high-dimensional problems easily.
