#### 数据格式处理 / Data Format Processing

在下面的例子中，我们将对DataFrame对象进行一些转换。

In the following example, we will perform some transformations on the DataFrame object.

apply方法是一个对列进行转换的强大工具。在把一个函数传给某列的apply方法之后，该函数将作用于该列中每个元素，生成一个新的Series对象并返回。

The `apply` method is a powerful tool for transforming a column. After passing a function to a column's `apply` method, that function will be applied to each element in the column, producing a new Series object.

这里我们将传递一个函数，去掉Population列字符串中的“,”，并将其类型转换为int。

Here we pass a function that removes the `,` characters from strings in the Population column and converts the type to int.

<!-- bilingual -->

```python
def remove_comma(x):
    if type(x) == str:
        x = x.replace(",", "")
    return int(x)

df_pop["Population"] = df_pop.Population.apply(remove_comma)
df_pop.head()
```

```python
df_pop.info()
```

Polulation列转换为了整数类型。

The Population column has been converted to integer type.

<!-- bilingual -->

```python
df_pop["State"][2]
```

我们还需要使用`strip`方法，去掉State和City列中每个元素末尾的空格。

We also need to use the `strip` method to remove trailing whitespace from each element in the State and City columns.

<!-- bilingual -->

```python
df_pop["State"] = df_pop.State.apply(lambda x: x.strip())
df_pop["City"] = df_pop.City.apply(lambda x: x.strip())
df_pop["State"][2]
```

In Pandas, there are dedicated data structures to store such data. Both Series and DataFrame can contain columns and indices of timestamps or time intervals.

Using Pandas' time indices `DatetimeIndex` and `PeriodIndex`, you can perform many date-, time-, and period-related operations.

#### 温度时间序列示例 / Temperature Time-Series Example

我们这里有两个数据文件 temperature_indoor_2014.tsv 和 temperature_indoor_2014.tsv。TSV格式是CSV格式的一种变体，每列数据之间使用Tab符号分隔。

Here we have two data files, temperature_indoor_2014.tsv and temperature_outdoor_2014.tsv. TSV is a variant of CSV in which columns are separated by Tab characters.

<!-- bilingual -->

```python
!head -n 5 temperature_outdoor_2014.tsv
```

使用`read_csv`函数读取数据文件，需要指定列之间的分隔符为Tab符号。因为本例文件中数据没有列头，我们显式地设置name关键词参数来命名列。

To read the data files with `read_csv`, the column delimiter must be specified as the Tab character. Because the files have no header, we explicitly set the `names` keyword argument to name the columns.

<!-- bilingual -->

```python
df1 = pd.read_csv('temperature_outdoor_2014.tsv', delimiter="\t", names=["time", "outdoor"])
df1.head()
```

```python
df2 = pd.read_csv('temperature_indoor_2014.tsv', delimiter="\t", names=["time", "indoor"])
df2.head()
```

下一步是把[UNIX时间戳](https://zh.m.wikipedia.org/zh/UNIX时间)（UTC1970年1月1日0时0分0秒起至现在的总秒数）转换为日期和时间对象。可以使用`to_datetime`方法，并设置参数unit="s"。

The next step is to convert the [UNIX timestamps](https://en.wikipedia.org/wiki/Unix_time) (the total seconds since 1970-01-01 00:00:00 UTC) into date-and-time objects. We can use the `to_datetime` method with the argument `unit="s"`.

<!-- bilingual -->

```python
df1.time = pd.to_datetime(df1.time.values, unit="s").tz_localize('UTC').tz_convert('Asia/Shanghai')
df1.time.head()
```

#### 重采样 / Resampling

Pandas时间序列对象的一个非常有用的功能是可以使用`resample`方法对数据进行升采样（up-sampling）和降采样（down-sampling）。对于升采样，需要用插值的方法填充缺失值。对于降采样，需要将多个数据点聚合成一个新的样本点。

A very useful feature of Pandas time-series objects is the ability to use the `resample` method for up-sampling and down-sampling. For up-sampling, missing values need to be filled via interpolation. For down-sampling, multiple data points need to be aggregated into a new sample point.

`resample`方法的第一个参数是一个字符串，用于指定重采样的频率。该方法会返回一个resampler对象，可以调用它的聚合方法，例如mean、sum等，获得重采样的数据。

The first argument of `resample` is a string specifying the resampling frequency. The method returns a resampler object on which aggregation methods such as `mean` or `sum` can be invoked to obtain the resampled data.

<!-- bilingual -->

我们示例中的原始数据的采样频率是10分钟，因为数据点较多，在绘图时需要进行降采样。

The original data in our example is sampled every 10 minutes. Because there are many data points, down-sampling is needed when plotting.

我们将使用四种不同的采样频率对温度时间序列进行重采样，并计算每天室内和室外平均温度的差值。

We will use four different sampling frequencies to resample the temperature time series and compute the daily difference between average indoor and outdoor temperatures.

<!-- bilingual -->

```python
df1_hour = df1.resample("H").mean()
df1_hour.columns = ["outdoor (hourly avg.)"]

df1_day = df1.resample("D").mean()
df1_day.columns = ["outdoor (daily avg.)"]

df1_week = df1.resample("7D").mean()
df1_week.columns = ["outdoor (weekly avg.)"]

df1_month = df1.resample("M").mean()
df1_month.columns = ["outdoor (monthly avg.)"]
```

```python
df_diff = (df1.resample("D").mean().outdoor - df2.resample("D").mean().indoor)
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 6))

df1_hour.plot(ax=ax1, alpha=0.25)
df1_day.plot(ax=ax1)
df1_week.plot(ax=ax1)
df1_month.plot(ax=ax1)

df_diff.plot(ax=ax2)
ax2.set_title("temperature difference between outdoor and indoor")

fig.tight_layout()
```

作为升采样的示例，我们使用三种不同的聚合方法（mean均值、ffill前值、bfill后值），对原始数据以5分钟的频率进行重采样。

As an example of up-sampling, we resample the original data at a 5-minute frequency using three different aggregation methods (`mean`, `ffill` for forward fill, and `bfill` for backward fill).

<!-- bilingual -->

```python
pd.concat([df1.resample("5min").mean().rename(columns={"outdoor": 'mean'}),
           df1.resample("5min").ffill().rename(columns={"outdoor": 'ffill'}),
           df1.resample("5min").bfill().rename(columns={"outdoor": 'bfill'})], axis=1).head()
```
