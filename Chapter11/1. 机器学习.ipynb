## 机器学习 / Machine Learning

---

机器学习是人工智能领域的一个分支。我们在这里简要介绍一下相关的概念和术语：

Machine learning is a branch of artificial intelligence. Here we briefly introduce some of the relevant concepts and terminology:

在机器学习中，将模拟或算法拟合到观测数据的过程称之为**训练 Traning**。

In machine learning, the process of fitting a model or algorithm to observed data is called **training**.

机器学习的应用通常可以分为两类：**监督学习 Supervised Learning**和**无监督学习 Unsupervised Learning**。

Applications of machine learning are often divided into two categories: **supervised learning** and **unsupervised learning**.

在监督学习中，数据包括**特征变量 Feature Variable**和已知的**响应变量 Independent Variable**。特征变量和相应变量可以是连续的或者离散的。通常需要利用相关的专业知识人工准备这些数据。典型的应用包括**回归 Regression**（对连续响应变量进行预测）和**分类 Classification**（对离散响应变量进行预测）。在**训练集 Traning Set**中，响应变量的值是已知的。

In supervised learning, the data include **feature variables** and known **response variables**. Feature and response variables can be continuous or discrete. Preparing such data usually requires domain knowledge and manual work. Typical applications include **regression** (predicting a continuous response) and **classification** (predicting a discrete response). In the **training set**, the values of the response variables are known.

无监督学习的应用是使用未标记或通过其它手动方式处理过的原始数据进行训练。无监督学习的典型应用是**聚类Clustering**（将数据分成合适的类别）。与监督学习相比，无监督学习的数据没有预先进行标记。这可能是由于样本数量太大，导致数据进行人工标记比较困难或者昂贵。显而易见，无监督学习更加困难，所受限制也较多。

Unsupervised learning is applied to raw data that are unlabeled or not otherwise manually processed. A typical use of unsupervised learning is **clustering** (grouping data into appropriate categories). Compared with supervised learning, unsupervised learning uses data that have not been pre-labeled. This is often because the sample size is so large that manual labeling is difficult or expensive. Unsupervised learning is obviously harder and more constrained.

<!-- bilingual -->

**交叉验证 Cross Validation**是将可用的数据分为训练集和测试集。训练集用于训练机器学习模型，而测试集则可以让训练后的模型对新数据进行测试。这样做的目的是检验模型对于新样本的预测能力，以及防止**过拟合 Over Fitting**。

**Cross-validation** splits the available data into a training set and a test set. The training set is used to train the machine-learning model, while the test set is used to evaluate the trained model on new data. The purpose is to assess the model's ability to predict new samples and to prevent **overfitting**.

有多种方法可以划分数据集用于交叉验证。**穷举交叉验证 Exhausive Cross-Validation**是测试所有的可能组合，不适用于大数据集；**留一交叉验证 Leave-one-out Cross-Validation**是每轮训练选择一个数据用于测试。这种方法的变体是将数据集分为k组，然后使用其中一组用于交叉验证。在scikit-learn库中，`sklearn.model_selection`模块提供了用于交叉验证的方法。

There are several ways to split the dataset for cross-validation. **Exhaustive cross-validation** tests all possible combinations and is not suitable for large datasets; **leave-one-out cross-validation** selects one sample for testing in each training round. A variant of this method splits the data into k groups and uses one of them for cross-validation. In scikit-learn, the `sklearn.model_selection` module provides methods for cross-validation.

**特征提取 Feature Extraction**是机器学习预处理截断的重要步骤，涉及创建合适的特征变量及对应的特征矩阵，这些特征矩阵可以传递给scikit-learn库中的某个机器学习算法。使用`sklearn.feature_extraction`模块可以从各种数据源中自动生成特征矩阵。

**Feature extraction** is an important step in machine-learning preprocessing; it involves creating appropriate feature variables and the corresponding feature matrix, which can be passed to a machine-learning algorithm in scikit-learn. The `sklearn.feature_extraction` module can automatically generate feature matrices from various data sources.

**降维 Dimensionality Reduction**和**特征选择 FeatureSelection**是机器学习中经常用到的方法。如果数据具有大量的解释变量（特征），但很多特征变量对于模型的预测能力影响很小。为了降低模型的复杂性，通常希望去除那些作用不太大的特征，从而减小问题的维度。scikit-learn库的`sklearn.decompositon`模块和`sklearn.feature_extraction`模块提供了各类降维方法。

**Dimensionality reduction** and **feature selection** are methods frequently used in machine learning. If the data have a large number of explanatory variables (features) but many of them contribute little to the model's predictive power, we usually want to remove those that are not very useful in order to reduce the complexity of the model and thus reduce the dimensionality of the problem. The `sklearn.decomposition` and `sklearn.feature_extraction` modules in scikit-learn provide various dimensionality-reduction methods.

## 回归 / Regression

---

回归是统计建模和机器学习的核心内容。

Regression is a core topic of statistical modeling and machine learning.

在机器学习中，相对于回归模型与已有数据的拟合成都，我们更关心的是模型对新样本的预测准确度。

In machine learning, compared with how well a regression model fits the available data, we care more about how accurately the model predicts new samples.

例如，如果有大量的特征数据和较少的观察样本，通常可以很完美地将回归模型拟合到数据，但这对于预测新数据没有什么作用，这就是典型地过拟合。

For example, with a large number of features and only a few observations, we can often fit a regression model perfectly to the data, but this is of little use for predicting new data — this is a typical case of overfitting.

为了提升模型的泛化能力，需要将可用数据划分为训练数据集和测试数据集。

To improve the model's generalization ability, the available data need to be split into a training set and a test set.

<!-- bilingual -->

下面，我们来看一个拥有50个样本以及50个特征的回归问题，其中只有10个特征具有有用信息。

Below we consider a regression problem with 50 samples and 50 features, of which only 10 features carry useful information.

<!-- bilingual -->

```python
np.random.seed(123)

X_all, y_all = datasets.make_regression(n_samples=50, n_features=50, 
                                        n_informative=10) #, noise=2.5)

X_all.shape, y_all.shape
```

得到的结果是两个数组，对应特征矩阵和样本。

The result consists of two arrays corresponding to the feature matrix and the samples.

为了避免过拟合，我们使用`model_selection`模块的`train_test_split`函数将数据集等分为训练集和测试集。

To avoid overfitting, we use the `train_test_split` function from the `model_selection` module to split the dataset evenly into a training set and a test set.

<!-- bilingual -->

```python
X_train, X_test, y_train, y_test = model_selection.train_test_split(X_all, y_all, train_size=0.5)

X_train.shape, y_train.shape
```

在sklearn中，可以使用`linear_model`模块的`LinearRegression`类来进行普通的线性回归。

In sklearn, the `LinearRegression` class in the `linear_model` module can be used for ordinary linear regression.

<!-- bilingual -->

```python
model = linear_model.LinearRegression()

model.fit(X_train, y_train)
```

`fit`方法会把结果直接保存在model实例中。

The `fit` method saves the results directly on the `model` instance.

<!-- bilingual -->

由于回归问题有50个特征，而训练数据只有25个样本，因此可以预期会发生过拟合。拟合程度可以通过模型与数据之间的误差平方和（SSE）来量化。

Since the regression problem has 50 features but the training data contain only 25 samples, overfitting is to be expected. The degree of fit can be quantified by the sum of squared errors (SSE) between the model and the data.

<!-- bilingual -->

```python
resid_train = y_train - model.predict(X_train)
sse_train = np.sum(resid_train**2)
sse_train
```

可以看到，拟合后的模型对于训练集数据的残差基本为0，可以预期过拟合的模型不适用于新数据。

As can be seen, the fitted model has essentially zero residuals on the training set, so the overfitted model can be expected to be unsuitable for new data.

<!-- bilingual -->

```python
resid_test = y_test - model.predict(X_test)
sse_test = np.sum(resid_test**2)
sse_test
```

评估模型和数据拟合度的另一个方法是R-squared统计量。如果结果是 0，说明模型拟合效果很差；如果结果是 1，说明模型无错误。

Another way to evaluate how well a model fits the data is the R-squared statistic. A value of 0 indicates a very poor fit, and a value of 1 indicates a perfect fit.

<!-- bilingual -->

```python
model.score(X_train, y_train), model.score(X_test, y_test)
```

最后，还可以使用图形方法对训练集和测试集的残差进行可视化。

Finally, we can visualize the residuals on the training and test sets using a graphical method.

<!-- bilingual -->

```python
def plot_residuals_and_coeff(resid_train, resid_test, coeff):
    fig, axes = plt.subplots(1, 3, figsize=(12, 3))
    axes[0].bar(np.arange(len(resid_train)), resid_train)
    axes[0].set_xlabel("sample number")
    axes[0].set_ylabel("residual")
    axes[0].set_title("training data")
    axes[1].bar(np.arange(len(resid_test)), resid_test)
    axes[1].set_xlabel("sample number")
    axes[1].set_ylabel("residual")
    axes[1].set_title("testing data")
    axes[2].bar(np.arange(len(coeff)), coeff)
    axes[2].set_xlabel("coefficient number")
    axes[2].set_ylabel("coefficient")
    fig.tight_layout()
    return fig, axes

fig, ax = plot_residuals_and_coeff(resid_train, resid_test, model.coef_)
```

在这里例子中，过拟合的原因是样本数量太少。

In this example, overfitting is caused by the small number of samples.

### 正则化回归 / Regularized Regression

解决过拟合问题的一个方案是**正则化回归 Regularized Regression**。

One solution to the overfitting problem is **regularized regression**.

在普通线性回归中，可以通过选择模型参数来最小化残差的平方和。优化问题的目标函数为$\min_\beta \Vert X \beta-y \Vert^2_2$，其中$X$是特征举证，$y$是响应变量，$\beta$是模型参数向量。

In ordinary linear regression, model parameters are chosen to minimize the sum of squared residuals. The optimization objective is $\min_\beta \Vert X \beta-y \Vert^2_2$, where $X$ is the feature matrix, $y$ is the response vector, and $\beta$ is the parameter vector.

在正则化回归中，我们在目标函数中增加了惩罚项，不同类型的惩罚项对原始回归问题进行不同的正则化。

In regularized regression, a penalty term is added to the objective function; different types of penalties impose different regularizations on the original regression problem.

最常见的正则化方法是在最小化目标函数中增加参数向量的L1范数和L2范数：$\min_\beta \left\{\Vert X \beta-y \Vert^2_2 + \alpha \Vert \beta \Vert_1  \right\}$和$\min_\beta \left\{\Vert X \beta-y \Vert^2_2 + \alpha \Vert \beta \Vert^2_2  \right\}$，这里$\alpha$是决定正则化强度的参数。它们分别被称为**LASSO回归**和**Ridge回归**。

The most common regularization adds an L1 norm or L2 norm of the parameter vector to the minimization objective: $\min_\beta \left\{\Vert X \beta-y \Vert^2_2 + \alpha \Vert \beta \Vert_1 \right\}$ and $\min_\beta \left\{\Vert X \beta-y \Vert^2_2 + \alpha \Vert \beta \Vert^2_2 \right\}$, where $\alpha$ determines the regularization strength. These are called **LASSO regression** and **Ridge regression**, respectively.

增加L1范数可以让模型参数向量减少非零项目，从而消除部分特征；增加L2范数可以让模型参数向量更小。

Adding the L1 norm tends to produce parameter vectors with fewer non-zero components, effectively removing some features; adding the L2 norm produces smaller parameter vectors overall.

在scikit-learn中，`linear_model`模块的`Ridge`类和`Lasso`类提供了相映的正则化回归方法。

In scikit-learn, the `Ridge` and `Lasso` classes in the `linear_model` module provide the corresponding regularized regression methods.

<!-- bilingual -->

```python
model = linear_model.Ridge(alpha=5)
model.fit(X_train, y_train)
```

计算模型对训练集和数据集的SSE。

Compute the SSE of the model on the training and test sets.

<!-- bilingual -->

```python
resid_train = y_train - model.predict(X_train)
sse_train = sum(resid_train**2)
sse_train
```

```python
resid_test = y_test - model.predict(X_test)
sse_test = sum(resid_test**2)
sse_test
```

可以看到，训练集的SSE不再是接近0的值，而测试集的SSE略有减小。

We can see that the SSE on the training set is no longer close to zero, and the SSE on the test set decreases slightly.

<!-- bilingual -->

```python
fig, ax = plot_residuals_and_coeff(resid_train, resid_test, model.coef_)
```

类型的，我们进行LASSO回归。在对实例进行初始化时，设置参数$\alpha=1.0$。

Similarly, we perform LASSO regression. When initializing the instance, we set the parameter $\alpha = 1.0$.

<!-- bilingual -->

```python
model = linear_model.Lasso(alpha=1.0)
model.fit(X_train, y_train)
```

```python
resid_train = y_train - model.predict(X_train)
sse_train = sum(resid_train**2)
sse_train
```

```python
resid_test = y_test - model.predict(X_test)
sse_test = sum(resid_test**2)
sse_test
```

可以看到，虽然训练集的SSE稍有增加，测试集的SSE明显下降。

We can see that, although the SSE on the training set has increased slightly, the SSE on the test set has decreased markedly.

<!-- bilingual -->

```python
fig, ax = plot_residuals_and_coeff(resid_train, resid_test, model.coef_)
```

在最初生成数据集时，我们选择了50个特征，但是只有10个特征包含有用信息。对于包含大量无用特征的模型，LASSO回归是一种非常有效的方法。

When the dataset was generated, we chose 50 features but only 10 carried useful information. For models with many useless features, LASSO regression is a very effective method.

### 正则化强度 / Regularization Strength

针对每个新问题，可能需要反复实验才能找到合适的$\alpha$值。

For each new problem, some trial and error may be required to find a suitable value of $\alpha$.

<!-- bilingual -->

```python
alphas = np.logspace(-4, 2, 11)
coeffs = np.zeros((len(alphas), X_train.shape[1]))
sse_train = np.zeros_like(alphas)
sse_test = np.zeros_like(alphas)
```

我们遍历数组中的$\alpha$值，并进行LASSO回归。

We iterate over the $\alpha$ values in the array and perform LASSO regression.

<!-- bilingual -->

```python
for n, alpha in enumerate(alphas):
    model = linear_model.Lasso(alpha=alpha)
    model.fit(X_train, y_train)
    coeffs[n, :] = model.coef_
    resid = y_train - model.predict(X_train)
    sse_train[n] = sum(resid**2)
    resid = y_test - model.predict(X_test)
    sse_test[n] = sum(resid**2)
```

对于较小的$\alpha$值，大量的系数都是非零的，这意味着过拟合。

For small $\alpha$ values, a large number of coefficients are non-zero, which indicates overfitting.

当$\alpha$值增加到某个阈值之上后，大部分系数变成0，同时测试集的SSE开始急速下降。着就是LASSO回归的从众效应 Sought-After Effect。

When $\alpha$ increases above a certain threshold, most of the coefficients become zero, and the SSE on the test set drops rapidly. This is the sought-after effect of LASSO regression.

但是当$\alpha$值太大时，所有系数都向0收敛，导致测试集和训练集的SSE开始上升。

However, when $\alpha$ becomes too large, all coefficients converge toward zero, causing the SSE on both training and test sets to rise.

<!-- bilingual -->

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharex=True)

for n in range(coeffs.shape[1]):
    axes[0].plot(np.log10(alphas), coeffs[:, n], color='k', lw=0.5)

axes[1].semilogy(np.log10(alphas), sse_train, label="train")
axes[1].semilogy(np.log10(alphas), sse_test, label="test")
axes[1].legend(loc=0)

axes[0].set_xlabel(r"${\log_{10}}\alpha$", fontsize=18)
axes[0].set_ylabel(r"coefficients", fontsize=18)
axes[1].set_xlabel(r"${\log_{10}}\alpha$", fontsize=18)
axes[1].set_ylabel(r"sse", fontsize=18)
fig.tight_layout()
```

我们可以使用`RidgeCV`和`LassoCV`类自动针对多个$\alpha$值测试正则化回归。它们能够在内部使用交叉验证的方法查找最优的$\alpha$值。

We can use the `RidgeCV` and `LassoCV` classes to automatically test regularized regression over multiple $\alpha$ values. They internally use cross-validation to find the optimal $\alpha$.

<!-- bilingual -->

```python
model = linear_model.LassoCV()
model.fit(X_all, y_all)
```

可以通过alpha_属性查看为模型选择的正则化强度参数。

The regularization strength chosen for the model can be inspected via the `alpha_` attribute.

<!-- bilingual -->

```python
model.alpha_
```

为了与之前的结果比较，我们重新计算训练集和测试集的SSE。

To compare with earlier results, we recompute the SSE on the training and test sets.

<!-- bilingual -->

```python
resid_train = y_train - model.predict(X_train)
sse_train = sum(resid_train**2)
sse_train
```

```python
resid_test = y_test - model.predict(X_test)
sse_test = sum(resid_test**2)
sse_test
```

```python
model.score(X_train, y_train), model.score(X_test, y_test)
```

```python
fig, ax = plot_residuals_and_coeff(resid_train, resid_test, model.coef_)
```

通过交叉验证的LASSO方法，我们获得了一个同时对训练集和测试集具有较高预测精度的模型，并且在样本数量少于特征数量的情况下解决了过拟合问题。

With cross-validated LASSO, we obtain a model with high predictive accuracy on both the training and test sets, and overfitting is resolved even when the number of samples is smaller than the number of features.

### Elastic-Net回归 / Elastic-Net Regression

Elastic-Net回归是结合了LASSO和Ridge回归中L1和L2范数的正则化回归模型。它的最小化目标函数是：
$$\min_\beta \left\{\Vert X \beta-y \Vert^2_2 + \alpha \rho \Vert \beta \Vert_1 +  \alpha (1 - \rho) \Vert \beta \Vert^2_2  \right\}$$
其中参数$\rho$是惩罚项的权重。

Elastic-Net regression is a regularized regression model that combines the L1 and L2 penalties of LASSO and Ridge regression. Its minimization objective is:
$$\min_\beta \left\{\Vert X \beta-y \Vert^2_2 + \alpha \rho \Vert \beta \Vert_1 +  \alpha (1 - \rho) \Vert \beta \Vert^2_2  \right\}$$
where the parameter $\rho$ is the weight of the L1 penalty term.

在scikit-learn中，ElasticNet类可以显式地设置参数$\alpha$和$\rho$，或者使用交叉验证的ElasticNetCV类自动寻找合适的参数值。

In scikit-learn, the `ElasticNet` class allows $\alpha$ and $\rho$ to be set explicitly, and `ElasticNetCV` can find suitable parameter values automatically via cross-validation.

<!-- bilingual -->

```python
model = linear_model.ElasticNetCV()
model.fit(X_all, y_all)
```

```python
model.alpha_, model.l1_ratio
```

```python
resid_train = y_train - model.predict(X_train)
sse_train = sum(resid_train**2)
sse_train
```

```python
resid_test = y_test - model.predict(X_test)
sse_test = sum(resid_test**2)
sse_test
```

```python
model.score(X_train, y_train), model.score(X_test, y_test)
```

```python
fig, ax = plot_residuals_and_coeff(resid_train, resid_test, model.coef_)
```

## 分类 / Classification

---

下面，我们将准备不同比例的训练集和测试集，测试和比较上述不同分类器的性能。

Below we prepare training and test sets of different proportions to test and compare the performance of the various classifiers.

<!-- bilingual -->

```python
train_size_vec = np.linspace(0.1, 0.9, 30)

classifiers = [tree.DecisionTreeClassifier,
               neighbors.KNeighborsClassifier,
               svm.SVC,
               ensemble.RandomForestClassifier
              ]

cm_diags = np.zeros((3, len(train_size_vec), len(classifiers)), dtype=float)
```

```python
for n, train_size in enumerate(train_size_vec):
    X_train, X_test, y_train, y_test = \
        model_selection.train_test_split(iris.data, iris.target, train_size=train_size)

    for m, Classifier in enumerate(classifiers): 
        classifier = Classifier()
        classifier.fit(X_train, y_train)
        y_test_pred = classifier.predict(X_test)
        cm_diags[:, n, m] = metrics.confusion_matrix(y_test, y_test_pred).diagonal()
        cm_diags[:, n, m] /= np.bincount(y_test)
```

```python
fig, axes = plt.subplots(1, len(classifiers), figsize=(12, 3))

for m, Classifier in enumerate(classifiers): 
    axes[m].plot(train_size_vec, cm_diags[2, :, m], label=iris.target_names[2])
    axes[m].plot(train_size_vec, cm_diags[1, :, m], label=iris.target_names[1])
    axes[m].plot(train_size_vec, cm_diags[0, :, m], label=iris.target_names[0])
    axes[m].set_title(type(Classifier()).__name__)
    axes[m].set_ylim(0, 1.1)
    axes[m].set_xlim(0.1, 0.9)
    axes[m].set_ylabel("classification accuracy")
    axes[m].set_xlabel("training size ratio")
    axes[m].legend(loc=4)

fig.tight_layout()
```

可以看到，每个模型的分类错误是不同的。针对特定问题，可能需要尝试不同的分类器。

It can be seen that the classification errors of each model differ. For a specific problem, one may need to try different classifiers.

## 聚类 / Clustering

---

聚类是类别未知的分类问题，它是无监督学习的典型应用。聚类算法的训练数据集仅包含特征变量，模型的输出是一个整数数组，代表为每个样本赋予的类别序号。

Clustering is the classification problem where the classes are unknown; it is a typical application of unsupervised learning. The training set for a clustering algorithm contains only the feature variables, and the model's output is an integer array giving the class index assigned to each sample.

scikit-learn库的`cluster`模块实现了多种聚类算法，例如k均值算法`KMeans`、mean-shift算法`MeanShift`等。很多聚类算法需要把簇的数量作为输入参数，可以在创建类实例时通过n_cluster参数来设置。

The `cluster` module of scikit-learn implements many clustering algorithms, such as `KMeans` for the k-means algorithm and `MeanShift` for the mean-shift algorithm. Many clustering algorithms take the number of clusters as an input parameter, which can be set via the `n_clusters` argument when instantiating the class.

这里，我们依旧使用Iris数据集来演示聚类算法。

Here we again use the Iris dataset to demonstrate clustering algorithms.

<!-- bilingual -->

```python
X, y = iris.data, iris.target
```

为了使用k均值算法，需要设置聚类的数目。通常情况下，需要尝试使用几个不同的聚类数量。

To use the k-means algorithm, the number of clusters must be specified. In general, it is helpful to try several different cluster counts.

<!-- bilingual -->

```python
n_clusters = 3
c = cluster.KMeans(n_clusters=n_clusters)
```

为了进行计算，我们调用聚合器的`fit`方法，这里只需要将特征矩阵作为参数输入。

To perform the computation, call the `fit` method of the clustering estimator; only the feature matrix is needed as input.

<!-- bilingual -->

```python
c.fit(X)
y_pred = c.predict(X)
y_pred 
```

```python
y
```

得到的结果是一个与训练集大小相同的整数数组，元素值代表分类组号。

The result is an integer array of the same size as the training set, whose element values represent cluster indices.

<!-- bilingual -->

```python
y_pred[::4]
```

可以对聚类的结果与监督分类的结果进行比较：

The clustering result can be compared with the result of supervised classification:

<!-- bilingual -->

```python
y[::4]
```

可以看到，二者的输出结果中类别的数字不同。为了对二者的指标进行比较，需要对数组的元素进行重命名。

It can be seen that the class numbers differ between the two outputs. To compare the metrics of the two, we need to rename the elements of the array.

<!-- bilingual -->

```python
idx_0, idx_1, idx_2 = (np.where(y_pred == n) for n in range(3))
y_pred[idx_0], y_pred[idx_1], y_pred[idx_2] = 1, 0, 2
y_pred
```

```python
np.sum(y_pred == y)
```

现在我们可以使用`confusion_matrix`方法比较监督分类和无监督分类的差别：

Now we can use the `confusion_matrix` function to compare supervised classification with unsupervised classification:

<!-- bilingual -->

```python
metrics.confusion_matrix(y, y_pred)
```

聚类算法可以正确识别第一类的样本，但是第二类有2个样本被划分到了第三类，第三类有14个样本被划分为了第二类。

The clustering algorithm correctly identifies the first class; however, 2 samples of the second class are assigned to the third class, and 14 samples of the third class are assigned to the second class.

<!-- bilingual -->

下面我们将对聚类的结果进行可视化。我们将遍历每个特征对以及每个聚类，并使用红色方框标记与监督分类结果不一致的样本。

Next we will visualize the clustering results. We iterate over each pair of features and each cluster, and use red squares to mark samples whose labels disagree with the supervised classification result.

<!-- bilingual -->

```python
N = X.shape[1]

fig, axes = plt.subplots(N, N, figsize=(12, 12), sharex=True, sharey=True)

colors = ["coral", "blue", "green"]
markers = ["^", "v", "o"]
for m in range(N):
    for n in range(N):
        for p in range(n_clusters):
            mask = y_pred == p
            axes[m, n].scatter(X[:, m][mask], X[:, n][mask],
                               marker=markers[p], s=30, 
                               color=colors[p], alpha=0.25)

        for idx in np.where(y != y_pred):
            axes[m, n].scatter(X[idx, m], X[idx, n],
                               marker="s", s=30, 
                               edgecolor="red", 
                               facecolor=(1,1,1,0))

    axes[N-1, m].set_xlabel(iris.feature_names[m], fontsize=16)
    axes[m, 0].set_ylabel(iris.feature_names[m], fontsize=16)
fig.tight_layout()
```

我们不能期望无监督聚类算法能够完美分类数据集。

We should not expect unsupervised clustering algorithms to perfectly classify a dataset.
