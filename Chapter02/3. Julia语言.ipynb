## 与Python的语法区别 / Syntactic differences from Python

* 对数组、字符串等索引。Julia 索引的下标是从 1 开始，而不是从 0 开始
* 索引列表和数组的最后一个元素时，Julia 使用 end ，Python 使用 -1
* Julia 中的 Comprehensions （还）没有条件 if 语句
* for, if, while, 等块的结尾需要 end ；不强制要求缩进排版
* Julia 没有代码分行的语法：如果在一行的结尾，输入已经是个完整的表达式，就直接执行；否则就继续等待输入。强迫 Julia 的表达式分行的方法是用圆括号括起来
* Julia 总是以列为主序的（类似 Fortran ），而 numpy 数组默认是以行为主序的（类似 C ）。如果想优化遍历数组的性能，从 numpy 到 Julia 时应改变遍历的顺序。

* Indexing into arrays, strings, etc. Julia indices start at 1, not at 0.
* When indexing the last element of a list or array, Julia uses end, while Python uses -1.
* Julia's comprehensions do not (yet) have a conditional if clause.
* Blocks such as for, if, and while require end at the end; indentation is not mandatory.
* Julia has no line-continuation syntax: if an input is a complete expression at the end of a line, it is executed immediately; otherwise the REPL waits for more input. To force an expression in Julia to span lines, wrap it in parentheses.
* Julia is always column-major (like Fortran), while NumPy arrays are row-major by default (like C). If you want to optimize the performance of iterating over arrays, change the iteration order when moving from NumPy to Julia.

### Julia 中的鸭子类型（Duck-typing） / Duck-typing in Julia

> *“假如它嘎嘎像个鸭鸭，那它就是个鸭鸭”*

> *"If it quacks like a duck, it's a duck"*

在未指定函数参数的类型时，不管输入参数是什么类型，只要函数体可以正常执行，
那么 Julia 函数就可以应用在这种类型的参数上  
比如上面的 `f` 函数可以接受一个矩阵（matrix）作为参数

When the type of a function argument is not specified, regardless of the type of input argument, as long as the function body can execute normally,
the Julia function can be applied to arguments of that type.  
For example, the `f` function above can accept a matrix as an argument.

<!-- bilingual -->

````julia
A = rand(3, 3)
````

````julia
f(A)
````

然而，`f` 函数无法接受一个向量（vector）。
不同于 `A^2` 是已经定义好的矩阵乘方，当 `v` 是向量时的 `v^2` 是未定义的，会导致错误。

However, the `f` function cannot accept a vector.
Unlike `A^2`, which is a well-defined matrix power, `v^2` when `v` is a vector is undefined and will cause an error.

<!-- bilingual -->

````julia
v = rand(3)
````

````julia
try
    f(v)
catch e
    println("MethodError: no method matching")
end
````

### 一些高阶函数 / Some higher-order functions

#### map函数 / The map function

`map` 是 Julia 中的高阶函数。“高阶”指的是它会*接受一个函数*作为它的传入参数。
`map` 将传入的函数作用于传入的数据结构的每一个元素。例如执行

`map` is a higher-order function in Julia. "Higher-order" means that it *takes a function* as one of its arguments.
`map` applies the passed-in function to every element of the passed-in data structure. For example, executing

```julia
map(f, [1, 2, 3])
```

将得到一个输出数组，数组的每一个元素为 `f` 作用在数组 `[1, 2, 3]` 的对应元素

yields an output array in which each element is the result of applying `f` to the corresponding element of the array `[1, 2, 3]`

```julia
[f(1), f(2), f(3)]
```

<!-- bilingual -->

````julia
map(f, [1, 2, 3])
````

这里我们把向量 `[1, 2, 3]` 的每一个元素都平方了，而不是对向量 `[1, 2, 3]` 本身进行平方。

Here we have squared each element of the vector `[1, 2, 3]`, rather than squaring the vector `[1, 2, 3]` itself.

`map` 可以接受匿名函数作为参数，像这样

`map` can take an anonymous function as an argument, like this

<!-- bilingual -->

````julia
map(x -> x^3, [1, 2, 3])
````

这就完成了对数组 `[1, 2, 3]` 所有元素立方的计算！

This is how we compute the cubes of all elements of the array `[1, 2, 3]`!

### 广播（broadcast） / Broadcasting

广播（`broadcast`）和 `map` 一样也是高阶函数。
广播是一种广义化的 `map`，所以 `map` 能做的它都能做。并不止如此，广播的调用方法和 `map` 一样。

Broadcasting (`broadcast`) is also a higher-order function, like `map`.
Broadcasting is a generalization of `map`, so anything `map` can do it can do as well. Moreover, broadcasting is called the same way as `map`.

<!-- bilingual -->

````julia
broadcast(f, [1, 2, 3])
````

我们又一次通过函数 `f` 把向量 `[1, 2, 3]` 的每一个元素都平方了
——这次是用“广播”函数 `f` 的方式！

Once again, we have squared each element of the vector `[1, 2, 3]` using the function `f`
— this time by "broadcasting" the function `f`!

广播（`broadcast`）有个语法糖：调用要广播的函数时，在函数名和输入变量之间加一个 `.`。

Broadcasting (`broadcast`) has syntactic sugar: when calling the function to broadcast, insert a `.` between the function name and the input variable.

比如：

For example:

```julia

```

就等价于

is equivalent to

```julia

```

````julia
f.([1, 2, 3])
````

We can compute the square of each element of a vector, but we cannot compute the square of the vector itself!

## 特殊的矩阵结构 / Special matrix structures

矩阵结构在线性代数中非常重要。

Matrix structure is very important in linear algebra.

用线性代数标准包 `LinearAlgebra` 可以获得结构化的矩阵（structured matrices）：

With the standard linear algebra package `LinearAlgebra`, we can obtain structured matrices:

<!-- bilingual -->

````julia
using LinearAlgebra
````

````julia
n = 1000
A = randn(n,n);
````

Julia 可以推断特殊矩阵结构，比如判断对称矩阵

Julia can infer special matrix structures, such as detecting symmetric matrices.

<!-- bilingual -->

````julia
Asym = A + A'
issymmetric(Asym)
````

但有时候浮点错误会比较麻烦

But sometimes floating-point errors can cause trouble.

<!-- bilingual -->

````julia
Asym_noisy = copy(Asym)
Asym_noisy[1,2] += 5eps()
````

````julia
issymmetric(Asym_noisy)
````

幸运的是我们可以通过如 `Diagonal`，`Triangular`，`Symmetric`，`Hermitian`，`Tridiagonal` 
和 `SymTridiagonal` 这样的函数来明确地创建矩阵

<!-- bilingual -->

Let us see how long Julia takes to compute the eigenvalues of `Asym`, `Asym_noisy`, and `Asym_explicit`.

<!-- bilingual -->

````julia
@time eigvals(Asym);
````

````julia
@time eigvals(Asym_noisy);
````

````julia
@time eigvals(Asym_explicit);
````

本例中，使用 `Symmetric()` 处理 `Asym_noisy` 后让计算效率提高了约5倍

In this example, processing `Asym_noisy` with `Symmetric()` made the computation roughly 5 times faster.

原作者不会 C 语言，所以他看不懂下面这个代码块中的代码，

The original author does not know C, so they cannot read the code in the block below,
but they know that C code can be compiled and run in Julia. Note that strings surrounded by `"""` are multi-line strings.

<!-- bilingual -->

````julia
using Libdl
C_code = """
#include <stddef.h>
double c_sum(size_t n, double *X) {
    double s = 0.0;
    for (size_t i = 0; i < n; ++i) {
        s += X[i];
    }
    return s;
}
"""

const Clib = tempname()  # 创建一个临时文件


# 通过将 C_code 导入 gcc 来编译成共享链接库
# (只有装了 gcc 才行):

open(`gcc -fPIC -O3 -msse3 -xc -shared -o $(Clib * "." * Libdl.dlext) -`, "w") do f
    print(f, C_code) 
end

# 定义一个调用这个 C 语言函数的 Julia 函数：
c_sum(X::Array{Float64}) = ccall(("c_sum", Clib), Float64, (Csize_t, Ptr{Float64}), length(X), X)
````

````julia
c_sum(a)
````

````julia
c_sum(a) ≈ sum(a)  # 输入\approx后按<TAB>键得到≈符号
````

````julia
c_sum(a) - sum(a)  
````

````julia
≈  # `isapprox` 函数的别名
````

现在我们可以在 Julia 中直接对 C 语言代码进行基准测试了：

Now we can benchmark C code directly from Julia:

<!-- bilingual -->

````julia
c_bench = @benchmark c_sum($a)
println("C: 最快 $(minimum(c_bench.times) / 1e6) 毫秒")
````

````julia
d = Dict()  # 字典，也就是关联数组（键值对）
d["C"] = minimum(c_bench.times) / 1e6  # 毫秒
````

````julia
function mysum_simd(A)   
    s = 0.0  # s = zero(eltype(A))
    @simd for a in A
        s += a
    end
    s
end
````

````julia
j_bench_hand_simd = @benchmark mysum_simd($a)
````

````julia
d["Julia hand-written simd"] = minimum(j_bench_hand_simd.times) / 1e6
d
````
