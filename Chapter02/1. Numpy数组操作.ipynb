## 内存中数组数据的顺序 / Order of array data in memory

---

对于包含行列的二维数据：一种可能的方式是按顺序逐行存储数据；一种可能的方式是按列存储数据。前者为行主序（row-major），后者为列主序（column-major）。C语言使用行主序，Fortran和MATLAB使用列主序。
NumPy数组默认使用行主序，创建数组时可以使用关键词参数`order='C'`或者`order='F'`设置。当使用借助C或者Fortran编写的接口时，需要特别注意NumPy数组的排序方式。

For two-dimensional data with rows and columns, one possible way is to store the data row by row in sequence; another is to store it column by column. The former is row-major, and the latter is column-major. C uses row-major order, while Fortran and MATLAB use column-major order.
NumPy arrays use row-major order by default. When creating an array, you can set this with the keyword argument `order='C'` or `order='F'`. When using interfaces written in C or Fortran, pay particular attention to the ordering of NumPy arrays.

### Meshgrid数组 / Meshgrid arrays

## 数组索引和切片 / Array indexing and slicing

---

正整数用于从数组开头进行索引（索引开始于0），负整数用于从数组结尾进行索引（索引开始于-1）。

Positive integers are used to index from the start of the array (indexing starts at 0), and negative integers are used to index from the end of the array (indexing starts at -1).

| 函数名 | 说明 |
|---|---|
| a[m] | 选择索引m处的元素，m为整数（从0开始计数） |
| a[-m] | 从数组末尾选择第m个元素，m为负整数（索引开始于-1） |
| a[m:n] | 选择索引为m到n-1的元素 |
| a[:] | 指定维的所有元素，等价为a[0:-1] |
| a[m:n:p] | 选择索引为m到n（不含）的间隔为p的元素 |
| a[::-1] | 逆序选择所有元素 |

| Function | Description |
|---|---|
| a[m] | Select the element at index m, where m is an integer (counting from 0) |
| a[-m] | Select the m-th element from the end of the array, where m is a negative integer (indexing starts at -1) |
| a[m:n] | Select elements from index m to n-1 |
| a[:] | All elements along the specified dimension, equivalent to a[0:-1] |
| a[m:n:p] | Select elements from index m to n (exclusive) with stride p |
| a[::-1] | Select all elements in reverse order |

<!-- bilingual -->

```python
a = np.arange(0, 11)
print(a)
print(a[5:])
print(a[::-2])
```

### 数组视图 / Array views

为了提高性能，使用切片操作提取的子数组是内存中同一组数据的不同视图。如果更改视图中的数据，原始数组中的数据也会被修改。如果需要数据的副本，可以使用`np.copy()`函数。

To improve performance, subarrays extracted by slicing operations are different views of the same data in memory. If you change data in the view, the data in the original array is also modified. If you need a copy of the data, use the `np.copy()` function.

<!-- bilingual -->

```python
b = a[::2, ::2]
print(b)
b[:, :] = 0
print(a)
```

## 向量化表达式 / Vectorized expressions

---

对标量和数组进行操作，广播会对标量与数组中的每个元素分别进行运算。当两个数组大小不同时，较小的数组通过广播匹配较大的数组。

When operating on scalars and arrays, broadcasting applies the operation between the scalar and each element of the array. When two arrays have different sizes, the smaller array is broadcast to match the larger one.

<!-- bilingual -->

```python
x = np.arange(9).reshape(3,3)
x
```

```python
y = np.arange(3)
x + y
```

```python
x + y.reshape(3,1)
```

对不兼容大小或形状的数组进行算数运算，则会引发ValueError错误。

Performing arithmetic operations on arrays with incompatible sizes or shapes raises a ValueError.

<!-- bilingual -->

```python
x = np.arange(4).reshape(2,2)
y = np.array([1,2,3])
print(x)
print(y)
try:
    print(x / y)
except ValueError:
    print("非法输入")
```

### 布尔数组和条件表达式 / Boolean arrays and conditional expressions

布尔数组和数值数组一起出现时，布尔数组会被转换为数值数组,False和True分别被转换为0和1。

When a boolean array appears together with a numeric array, the boolean array is converted to a numeric array: False and True are converted to 0 and 1, respectively.

<!-- bilingual -->

```python
x = np.linspace(-3,3,7)
x * (x > 0)
```

| 函数名 | 说明 |
|---|---|
| np.where  | 根据条件数组的值从连个数组中选择值 |

<!-- bilingual -->

```python
np.where(x < 0)  # 当where内只有一个参数时，返回一个元组
```

```python
np.where(x < 0, x**2, x**3)  # 当where内有三个参数时，第一个参数表示条件，选择性返回后两个数组
```

```python
np.select([x < -1, x > 1], [x**2, x**3])  # 根据条件列表从数组列表中选择值
```

```python
np.choose([0, 0, 1, 1, 0, 0, 1], [x**2, x**3]) # choose 第一个参数中的索引决定了从数组列表中哪一个选择值
```

```python
x[np.nonzero(x)]  # x[x != 0] 
```

```python
A = np.arange(1, 7).reshape(2, 3)
B = np.arange(1, 7).reshape(3, 2)
print(A)
print(B)

np.dot(A, B) # @符号也可以用来表示矩阵乘法，A@B
```

NumPy为ndarray提供另一种数据结构matrix，诸如`A * B`之类的表达式可以使用matrix乘法表示。逆矩阵表示为 `matrix.I`，复共轭转置表示为 `matrix.H`。与`np.matrix`创建数据副本不同，`np.asmatrix`函数以`np.matrix`实例的形式创建原始数组的视图，不增加计算成本。

In addition to ndarray, NumPy provides another data structure, matrix, where expressions such as `A * B` can be written using matrix multiplication. The inverse matrix is written as `matrix.I`, and the conjugate transpose as `matrix.H`. Unlike `np.matrix`, which creates a copy of the data, the `np.asmatrix` function creates a view of the original array as an `np.matrix` instance with no additional computational cost.

<!-- bilingual -->

```python
A = np.matrix(A)
B = np.matrix(B)
A * B
```

```python
np.array(A * B)  # 类似的，np.array将创建数据副本，np.asarray将创建数据视图
```
