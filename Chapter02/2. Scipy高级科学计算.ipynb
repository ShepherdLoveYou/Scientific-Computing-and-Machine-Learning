SciPy包含许多专注于科学计算中的常见问题的工具箱。它的子模块对应于不同的应用，比如插值、积分、优化、图像处理、统计和特殊功能等。

SciPy contains many toolboxes focused on common problems in scientific computing. Its submodules correspond to different applications such as interpolation, integration, optimization, image processing, statistics, and special functions.

SciPy可以与其他标准科学计算包相对比，比如GSL (C和C++的GNU科学计算包), 或者Matlab的工具箱。SciPy是Python中科学程序的核心程序包；这意味着有效的操作NumPy数组，因此，NumPy和SciPy可以一起工作。

SciPy can be compared to other standard scientific computing packages, such as GSL (the GNU Scientific Library for C and C++), or Matlab's toolboxes. SciPy is the core package for scientific programs in Python; this means efficient manipulation of NumPy arrays, so NumPy and SciPy work well together.

在实现一个程序前，有必要确认一下需要的数据处理方式是否已经在SciPy中实现了。作为非专业程序员，科学家通常倾向于**重新发明轮子**，这产生了小玩具、不优化、很难分享以及不可以维护的代码。相反，SciPy的程序是优化并且测试过的，因此应该尽可能使用。

Before implementing a routine, it is worth checking whether the desired data processing has already been implemented in SciPy. As non-professional programmers, scientists often tend to **reinvent the wheel**, which results in buggy, non-optimal, hard-to-share, and unmaintainable code. By contrast, SciPy's routines are optimized and tested, and should therefore be used whenever possible.

**警告** 这个教程根本不是数值计算的介绍。因为列举SciPy的不同子模块和功能将会是非常枯燥的，相反我们将聚焦于列出一些例子，给出如何用SciPy进行科学计算的大概思路。

The [scipy.linalg](http://docs.scipy.org/doc/scipy/reference/linalg.html#scipy.linalg) module provides standard linear algebra operations, relying on efficient underlying implementations (BLAS, LAPACK).

* [scipy.linalg.det()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.linalg.det.html#scipy.linalg.det) 函数计算方阵的行列式：

* The [scipy.linalg.det()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.linalg.det.html#scipy.linalg.det) function computes the determinant of a square matrix:

<!-- bilingual -->

```python
from scipy import linalg
arr = np.array([[1, 2],
                [3, 4]])
linalg.det(arr)
```

```python
arr = np.array([[3, 2],
                 [6, 4]])
linalg.det(arr)
```

```python
try:
    linalg.det(np.ones((3, 4)))
except ValueError:
    print("expected square matrix")
```

* [scipy.linalg.inv()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.linalg.inv.html#scipy.linalg.inv) 函数计算逆方阵：

* The [scipy.linalg.inv()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.linalg.inv.html#scipy.linalg.inv) function computes the inverse of a square matrix:

<!-- bilingual -->

```python
arr = np.array([[1, 2],
                 [3, 4]])
iarr = linalg.inv(arr)
iarr
```

```python
np.allclose(np.dot(arr, iarr), np.eye(2))
```

最后计算逆奇异矩阵（行列式为0）将抛出`LinAlgError` :

Finally, computing the inverse of a singular matrix (determinant equal to 0) will throw a `LinAlgError`:

<!-- bilingual -->

```python
arr = np.array([[3, 2],
                [6, 4]])

try:
    linalg.inv(arr)
except np.linalg.LinAlgError as err:
    print("singular matrix")
```

* 还有更多高级的操作，奇异值分解（SVD）：

* There are more advanced operations as well, such as singular value decomposition (SVD):

<!-- bilingual -->

```python
arr = np.arange(9).reshape((3, 3)) + np.diag([1, 0, 1])
uarr, spec, vharr = linalg.svd(arr)
```

结果的数组频谱是：

The resulting array of spectra is:

<!-- bilingual -->

```python
spec
```

原始矩阵可以用`svd`和`np.dot`矩阵相乘的结果重新获得：

The original matrix can be recovered using `svd` and `np.dot` matrix multiplication:

<!-- bilingual -->

```python
sarr = np.diag(spec)
svd_mat = uarr.dot(sarr).dot(vharr)
np.allclose(svd_mat, arr)
```

SVD常被用于统计和信号处理。其他标准分解 (QR, LU, Cholesky, Schur), 以及线性系统的求解器，也可以在[scipy.linalg](http://docs.scipy.org/doc/scipy/reference/linalg.html#scipy.linalg)中找到。

SVD is often used in statistics and signal processing. Other standard decompositions (QR, LU, Cholesky, Schur), as well as solvers for linear systems, can also be found in [scipy.linalg](http://docs.scipy.org/doc/scipy/reference/linalg.html#scipy.linalg).

## 快速傅立叶变换：[scipy.fftpack](http://docs.scipy.org/doc/scipy/reference/fftpack.html#scipy.fftpack) / Fast Fourier transforms: [scipy.fftpack](http://docs.scipy.org/doc/scipy/reference/fftpack.html#scipy.fftpack)

---

[scipy.fftpack](http://docs.scipy.org/doc/scipy/reference/fftpack.html#scipy.fftpack) 模块允许计算快速傅立叶变换。例子，一个（有噪音）的信号输入是这样：

The [scipy.fftpack](http://docs.scipy.org/doc/scipy/reference/fftpack.html#scipy.fftpack) module allows you to compute fast Fourier transforms. For example, a (noisy) signal input looks like this:

<!-- bilingual -->

```python
time_step = 0.02
period = 5.
time_vec = np.arange(0, 20, time_step)
sig = np.sin(2 * np.pi / period * time_vec) + \
       0.5 * np.random.randn(time_vec.size)

import matplotlib.pyplot as plt
plt.figure(figsize=(6, 5))
plt.plot(time_vec, sig, label='Original signal')
```

观察者并不知道信号的频率，只知道抽样时间步骤的信号`sig`。假设信号来自真实的函数，因此傅立叶变换将是对称的。[scipy.fftpack.fftfreq()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.fftpack.fftfreq.html#scipy.fftpack.fftfreq) 函数将生成样本序列，而[scipy.fftpack.fft()]()将计算快速傅立叶变换：

The observer does not know the frequency of the signal, only the signal `sig` at the sampled time steps. Assume the signal comes from a real-valued function, so the Fourier transform will be symmetric. The [scipy.fftpack.fftfreq()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.fftpack.fftfreq.html#scipy.fftpack.fftfreq) function generates the sample sequence, and [scipy.fftpack.fft()]() computes the fast Fourier transform:

<!-- bilingual -->

```python
from scipy import fftpack
sample_freq = fftpack.fftfreq(sig.size, d=time_step)
sig_fft = fftpack.fft(sig)
```

因为生成的幂是对称的，寻找频率只需要使用频谱为正的部分：

Because the resulting power is symmetric, finding the frequency only requires using the positive part of the spectrum:

<!-- bilingual -->

```python
pidxs = np.where(sample_freq > 0)
freqs = sample_freq[pidxs]
power = np.abs(sig_fft)[pidxs]
```

寻找信号频率：

Find the signal frequency:

<!-- bilingual -->

```python
freq = freqs[power.argmax()]
np.allclose(freq, 1./period)  # 检查是否找到了正确的频率
```

现在高频噪音将从傅立叶转换过的信号移除：

Now the high-frequency noise will be removed from the Fourier-transformed signal:

<!-- bilingual -->

```python
sig_fft[np.abs(sample_freq) > freq] = 0
```

生成的过滤过的信号可以用[scipy.fftpack.ifft()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.fftpack.ifft.html#scipy.fftpack.ifft)函数：

The resulting filtered signal can be obtained with the [scipy.fftpack.ifft()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.fftpack.ifft.html#scipy.fftpack.ifft) function:

<!-- bilingual -->

```python
main_sig = fftpack.ifft(sig_fft)
```

查看结果：

View the result:

<!-- bilingual -->

```python
plt.figure(figsize=(6, 5))
plt.plot(time_vec, sig)
plt.plot(time_vec, main_sig.real, linewidth=3)
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.show()
```

### 寻找标量函数的最小值 / Finding the minimum of a scalar function

让我们定义下面的函数：

Let us define the following function:

<!-- bilingual -->

```python
def f(x):
    return x**2 + 10*np.sin(x)
```

绘制它：

Plot it:

<!-- bilingual -->

```python
x = np.arange(-10, 10, 0.1)
plt.plot(x, f(x)) 
plt.show() 
```

这个函数在-1.3附近有一个全局最小并且在3.8有一个局部最小。

This function has a global minimum near -1.3 and a local minimum at 3.8.

找到这个函数的最小值的常用有效方式是从给定的初始点开始进行一个梯度下降。[BFGS算法](https://zh.m.wikipedia.org/wiki/BFGS%E7%AE%97%E6%B3%95)是这样做的较好方式：

A common and efficient way to find the minimum of this function is to perform gradient descent from a given initial point. The [BFGS algorithm](https://zh.m.wikipedia.org/wiki/BFGS%E7%AE%97%E6%B3%95) is a good way to do this:

<!-- bilingual -->

```python
optimize.fmin_bfgs(f, 0)
```

这个方法的一个可能问题是，如果这个函数有一些局部最低点，算法可能找到这些局部最低点而不是全局最低点，这取决于初始点：

A possible issue with this method is that if the function has several local minima, the algorithm may find these local minima instead of the global minimum, depending on the initial point:

<!-- bilingual -->

```python
optimize.fmin_bfgs(f, 3, disp=0)
```

如果我们不知道全局最低点，并且使用其临近点来作为初始点，那么我们需要付出昂贵的代价来获得全局最优。要找到全局最优点，最简单的算法是暴力算法，算法中会评估给定网格内的每一个点：

If we do not know the global minimum and use one of its neighbors as the initial point, we need to pay a heavy price to obtain the global optimum. To find the global optimum, the simplest algorithm is brute force, in which every point within a given grid is evaluated:

<!-- bilingual -->

```python
grid = (-10, 10, 0.1)
xmin_global = optimize.brute(f, (grid,))
xmin_global
```

对于更大的网格，[scipy.optimize.brute()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.brute.html#scipy.optimize.brute) 变得非常慢。[scipy.optimize.anneal()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.anneal.html#scipy.optimize.anneal) 提供了一个替代的算法，使用模拟退火。对于不同类型的全局优化问题存在更多的高效算法，但是这超出了`scipy`的范畴。[OpenOpt](http://openopt.org/Welcome)、[IPOPT](https://github.com/xuy/pyipopt)、[PyGMO](http://pagmo.sourceforge.net/pygmo/index.html)和[PyEvolve](http://pyevolve.sourceforge.net/)是关于全局优化的一些有用的包。

For larger grids, [scipy.optimize.brute()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.brute.html#scipy.optimize.brute) becomes very slow. [scipy.optimize.anneal()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.anneal.html#scipy.optimize.anneal) provides an alternative algorithm that uses simulated annealing. More efficient algorithms exist for different kinds of global optimization problems, but these are beyond the scope of `scipy`. [OpenOpt](http://openopt.org/Welcome), [IPOPT](https://github.com/xuy/pyipopt), [PyGMO](http://pagmo.sourceforge.net/pygmo/index.html), and [PyEvolve](http://pyevolve.sourceforge.net/) are some useful packages for global optimization.

### 统计检验 / Statistical tests

统计检验是一个决策指示器。例如，如果我们有两组观察值，我们假设他们来自于高斯过程，我们可以用T检验来决定这两组观察值是不是显著不同：

生成的结果由以下内容组成：

The result consists of:

* T 统计值：一个值，符号与两个随机过程的差异成比例，大小与差异的程度有关。
* p 值：两个过程相同的概率。如果它接近1，那么这两个过程几乎肯定是相同的。越接近于0，越可能这两个过程有不同的平均数。

* T statistic: a value whose sign is proportional to the difference between the two random processes and whose magnitude is related to the degree of this difference.
* p-value: the probability that the two processes are the same. If it is close to 1, then the two processes are almost certainly identical. The closer it is to 0, the more likely the two processes have different means.

`odeint`解决如下形式的第一顺序ODE系统：

`odeint` solves systems of first-order ODEs of the form:

$dy/dt = rhs(y1, y2, .., t0,...)$

作为一个介绍，让我们解一下在初始条件下$y(t=0) = 1$，这个常微分公式$dy/dt = -2y$在$t = 0..4$时的值。首先，这个函数计算定义位置需要的导数：

As an introduction, let us solve, under the initial condition $y(t=0) = 1$, the ordinary differential equation $dy/dt = -2y$ on $t = 0..4$. First, the function computes the derivative needed at the definition location:

<!-- bilingual -->

```python
def calc_derivative(ypos, time, counter_arr):
    counter_arr += 1
    return -2 * ypos
```

添加了一个额外的参数`counter_arr`用来说明这个函数可以在一个时间步骤被调用多次，直到收敛。计数器数组定义如下：

An extra argument `counter_arr` is added to illustrate that this function may be called multiple times within a single time step, until convergence. The counter array is defined as follows:

<!-- bilingual -->

```python
counter = np.zeros((1,), dtype=np.uint16)
```

现在计算轨迹线：

Now compute the trajectory:

<!-- bilingual -->

```python
from scipy.integrate import odeint
time_vec = np.linspace(0, 4, 40)
yvec, info = odeint(calc_derivative, 1, time_vec,
                    args=(counter,), full_output=True)
```

因此，导数函数被调用了40多次（即时间步骤数）：

Thus, the derivative function was called more than 40 times (i.e. the number of time steps):

阻尼弹簧重物振子（二阶振荡器）是使用[scipy.integrate.odeint()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.odeint.html#scipy.integrate.odeint)的另一个例子。链接到弹簧的重物的位置服从二阶常微分方程$y'' + 2 eps wo  y' + wo^2 y = 0$，其中$wo^2 = k/m$ 弹簧的常数为k, m是重物质量，$eps=c/(2 m wo)$，c是阻尼系数。例如，我们选择如下参数：

A damped spring-mass oscillator (second-order oscillator) is another example of using [scipy.integrate.odeint()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.odeint.html#scipy.integrate.odeint). The position of a mass attached to a spring obeys the second-order ODE $y'' + 2 eps wo  y' + wo^2 y = 0$, where $wo^2 = k/m$, the spring constant is k, m is the mass, $eps=c/(2 m wo)$, and c is the damping coefficient. For example, we choose the following parameters:

<!-- bilingual -->

```python
mass = 0.5  # kg
kspring = 4  # N/m
cviscous = 0.4  # N s/m
```

因此系统将是欠阻尼的，因为：

Therefore the system will be underdamped, since:

<!-- bilingual -->

```python
eps = cviscous / (2 * mass * np.sqrt(kspring/mass))
eps < 1
```

对于[scipy.integrate.odeint()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.odeint.html#scipy.integrate.odeint)求解器，二阶等式需要被变换为系统内向量$Y=(y, y')$的两个一阶等式。为了方便，定义$nu = 2 eps * wo = c / m$和$om = wo^2 = k/m$：

For the [scipy.integrate.odeint()](http://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.odeint.html#scipy.integrate.odeint) solver, the second-order equation needs to be transformed into two first-order equations for the vector $Y=(y, y')$ within the system. For convenience, define $nu = 2 eps * wo = c / m$ and $om = wo^2 = k/m$:

<!-- bilingual -->

```python
nu_coef = cviscous / mass
om_coef = kspring / mass
```

因此函数将计算速度和加速度：

The function will therefore compute velocity and acceleration:

<!-- bilingual -->

```python
def calc_deri(yvec, time, nuc, omc):
    return (yvec[1], -nuc * yvec[1] - omc * yvec[0])

time_vec = np.linspace(0, 10, 100)
yarr = odeint(calc_deri, (1, 0), time_vec, args=(nu_coef, om_coef))
```

如下的Matplotlib图片显示了最终的位置和速度：

The following Matplotlib figure shows the resulting position and velocity:

<!-- bilingual -->

```python
plt.figure(figsize=(4, 3))
plt.plot(time_vec, yarr[:, 0], label='y')
plt.plot(time_vec, yarr[:, 1], label="y'")
plt.legend(loc='best')
```

### 图像滤波器 / Image filters

<!-- bilingual -->

```python
from scipy.datasets import face as _face

face = _face(gray=True)
face = face[:512, -512:]  # crop out square on right
import numpy as np
noisy_face = np.copy(face).astype(float)
noisy_face += face.std() * 0.5 * np.random.standard_normal(face.shape)
blurred_face = ndimage.gaussian_filter(noisy_face, sigma=3)
median_face = ndimage.median_filter(noisy_face, size=5)
```

```python
plt.figure(figsize=(9, 3.5))
plt.subplot(131)
plt.imshow(noisy_face, cmap=plt.cm.gray)
plt.axis('off')
plt.title('noisy')

plt.subplot(132)
plt.imshow(blurred_face, cmap=plt.cm.gray)
plt.axis('off')
plt.title('Gaussian filter')

plt.subplot(133)
plt.imshow(median_face, cmap=plt.cm.gray)
plt.axis('off')
plt.title('median filter')

plt.subplots_adjust(wspace=.05, left=.01, bottom=.01, right=.99, top=.99)
```

### 数学形态学 / Mathematical morphology

[数学形态学](https://en.wikipedia.org/wiki/Mathematical_morphology)是集合理论分支出来的一个数学理论。它刻画并转换几何结构。特别是二元的图像（黑白）可以用这种理论来转换：被转换的集合是临近非零值像素的集合。这个理论也可以被扩展到灰度值图像。

[Mathematical morphology](https://en.wikipedia.org/wiki/Mathematical_morphology) is a mathematical theory branching out from set theory. It characterizes and transforms geometric structures. In particular, binary images (black and white) can be transformed with this theory: the set to be transformed is the set of neighboring non-zero pixels. This theory can also be extended to grayscale images.

![](Images/morpho_mat.png)

Elementary mathematical morphology operations use a structuring element to modify other geometric structures.

开启操作移除小的结构，而关闭操作填满了小洞。因此这些用来”清洗“图像。

The opening operation removes small structures, while the closing operation fills small holes. These are therefore used to "clean" an image.

```python
a = np.zeros((50, 50))
a[10:-10, 10:-10] = 1
a += 0.25*np.random.standard_normal(a.shape)
mask = a>=0.5
opened_mask = ndimage.binary_opening(mask)
closed_mask = ndimage.binary_closing(opened_mask)
```
