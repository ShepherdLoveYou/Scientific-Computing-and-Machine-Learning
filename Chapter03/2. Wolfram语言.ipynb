# Wolfram语言 / Wolfram Language

---

<img src="https://upload.wikimedia.org/wikipedia/commons/thumb/e/eb/WolframCorporateLogo.svg/1920px-WolframCorporateLogo.svg.png" style="width: 20%;"/>

Wolfram 语言（Wolfram Language, Wolflang）是一门通用多范式计算语言，由 Wolfram Research 开发，被用于其旗下产品 Mathematica，Wolfram Cloud 之中。它的功能覆盖非常广泛，可以用于解决大量专业领域的问题。例如，它内置了用于生成和运行图灵机、创建图形和音频、分析三维模型、矩阵操作、求解微分方程的内置函数。

The Wolfram Language (Wolflang) is a general-purpose multi-paradigm computing language developed by Wolfram Research and used in its products such as Mathematica and Wolfram Cloud. Its capabilities span an extremely wide range and can be used to solve problems in many specialized domains. For example, it has built-in functions for generating and running Turing machines, creating graphics and audio, analyzing 3D models, manipulating matrices, and solving differential equations.

Wolfram 语言是内置计算智能的编程，其理念是在语言内部构建尽可能多的关于算法以及世界的知识。Wolfram 语言比 Python 拥有更高级别和更综合化的设计原则，基于完全的符号式语言，以及桌面和云端的无缝操作，和在语言中直接构建的世界最大的算法和数据集成，全部统一的设计和文档，可访问的世界独创笔记本界面，允许更快和更高级开发部署您的理念，无需查找并学习单独的程序库。

The Wolfram Language is a programming language with built-in computational intelligence; its philosophy is to build as much knowledge about algorithms and about the world as possible directly into the language. The Wolfram Language has higher-level and more integrated design principles than Python. It is based on a fully symbolic language, offers seamless desktop and cloud operation, has the world's largest algorithm and data integration built directly into the language, all with unified design and documentation, an accessible world-first notebook interface — allowing you to develop and deploy ideas faster and at a higher level, without hunting down and learning separate libraries.

Wolfram 语言是闭源的商业软件。但是Wolfram向个人开发者提供了免费 [Wolfram Engine](https://www.wolfram.com/engine-technology)。免费的开发者 Wolfram 引擎包含一个 Wolfram Cloud Basic 的许可证。Wolfram 引擎可以自动访问 Wolfram Knowledgebase 并下载任何所需资源。本Jupyter文档使用了[Wolfram Language kernel](https://github.com/WolframResearch/WolframLanguageForJupyter)。

The Wolfram Language is closed-source commercial software. However, Wolfram provides a free [Wolfram Engine](https://www.wolfram.com/engine-technology) for individual developers. The free developer Wolfram Engine includes a Wolfram Cloud Basic license. The Wolfram Engine can automatically access the Wolfram Knowledgebase and download any required resources. This Jupyter document uses the [Wolfram Language kernel](https://github.com/WolframResearch/WolframLanguageForJupyter).
