与数值运算不同，符号计算（Symbolic Computing）是一种完全不同的计算模式。在符号计算软件中（Computer Algebra System， CAS）中，数学对象和表达式用符号表示，以解析方式进行变换和计算。在使用数值计算之前，通过解析推导，可以降低最终待解决问题的复杂度和解的空间大小。

Unlike numerical computation, symbolic computation is a completely different mode of calculation. In a computer algebra system (CAS), mathematical objects and expressions are represented symbolically, and transformations and computations are performed analytically. By performing analytical derivation before numerical computation, one can reduce the complexity of the final problem and the size of the solution space.

在Python科学计算环境中，主要用到的符号计算库是SymPy。SymPy完全使用Python编写，提供了一系列用于解析计算和符号计算的工具。

In the Python scientific computing ecosystem, the main symbolic computation library is SymPy. SymPy is written entirely in Python and provides a set of tools for analytical and symbolic computation.

## 数值计算 / Numerical Evaluation

---

对符号表达式进行求值计算是一个常用操作。可以使用函数`sympy.N`或者表达式实例的`evalf`方法对SymPy表达式进行求值。

Evaluating symbolic expressions numerically is a common operation. You can use the `sympy.N` function or the `evalf` method of an expression instance to numerically evaluate SymPy expressions.

<!-- bilingual -->

```python
sympy.N(1 + sympy.pi)
```

```python
(1 + sympy.pi).evalf(5)  # 指定结果的有效位数
```

当对一系列输入值进行数值计算时，原则上可以对这些输入值进行循环，然后调用`evalf`。

When performing numerical computation over a series of input values, in principle you can loop over these input values and call `evalf`.

<!-- bilingual -->

```python
expr = sympy.sin(sympy.pi * x * sympy.exp(x))
[expr.subs(x, i).evalf(3) for i in range(0,5)]
```

我们可以使用更高效的函数`sympy.lambdify`执行上述操作。

We can use the more efficient function `sympy.lambdify` to perform the above operation.

<!-- bilingual -->

```python
expr_func = sympy.lambdify(x, expr)
expr_func(1.0)
```

需要注意，上述expr_func需要将数字作为参数，我们不能将符号直接作为参数转递给改函数。同时，上述expr_func也不兼容NumPy数组的矢量输入。为了产生NumPy数组兼容的函数，可以将'numpy'作为第三个参数传递给`sympy.lambdify`。

Note that the resulting expr_func must be called with numbers as arguments; we cannot pass symbols directly to this function. Additionally, this expr_func is not compatible with vectorized NumPy array inputs. To produce a function compatible with NumPy arrays, pass 'numpy' as the third argument to `sympy.lambdify`.

<!-- bilingual -->

```python
expr_func = sympy.lambdify(x, expr)
import numpy as np
xvalues = np.arange(0, 10)
expr_func(xvalues)
```

When this method is applied to explicit functions, the integrals will be computed accordingly.

### 级数展开 / Series Expansion

级数展开时很多计算领域的重要工具。通过级数展开，可以将任意函数写为一个多项式，其系数由函数的导数在展开处的点给出。

Series expansion is an important tool in many computational fields. Through series expansion, an arbitrary function can be written as a polynomial whose coefficients are given by the derivatives of the function evaluated at the expansion point.

在数学上，对于一个在实数或复数$a$邻域上，以实数作为变量或以复数作为变量的函数，并且是无穷可微的函数$f(x)$，它的泰勒级数是以下这种形式的幂级数：
$$\sum_{n=0}^{\infty} \frac{f^{(n)}(a)}{n!} (x-a)^{n}$$

Mathematically, for a function $f(x)$ that is infinitely differentiable in a neighborhood of a real or complex number $a$ with a real or complex variable, its Taylor series is a power series of the following form:
$$\sum_{n=0}^{\infty} \frac{f^{(n)}(a)}{n!} (x-a)^{n}$$

`sympy.series`函数的第一个参数是需要展开的函数或表达式，第二个参数是需要展开的符号。另外，还可以指定围绕哪个点展开（默认x0=0）和展开的阶数（默认n=6）。

The first argument of the `sympy.series` function is the function or expression to expand, and the second argument is the symbol to expand around. In addition, you can specify the point around which to expand (default x0=0) and the order of the expansion (default n=6).

<!-- bilingual -->

```python
x, y = sympy.symbols("x, y")
f = sympy.Function("f")(x)
sympy.series(f, x)
```

为了改变展开的点，可以设置x0参数。

To change the expansion point, set the x0 parameter.

<!-- bilingual -->

```python
x0 = sympy.Symbol("{x_0}")
f.series(x, x0, n=2)
```

截断产生的误差用高阶无穷小项$O(x-x_0)^n$表示。对于数值计算，需要从表达式中移除误差项。

The error caused by truncation is represented by a higher-order infinitesimal term $O(x-x_0)^n$. For numerical computation, this error term must be removed from the expression.

<!-- bilingual -->

```python
f.series(x, x0, n=2).removeO()
```

对于特定函数和表达式，级数展开可以得到计算结果。

For certain functions and expressions, the series expansion yields an explicit computed result.

<!-- bilingual -->

Equation solving is a fundamental application of mathematics. SymPy can symbolically solve various equations, but in principle many equations do not have an analytical solution. ## 微分方程 / Differential Equations

---

凡含有参数，未知函数和未知函数导数 (或微分) 的方程，称为微分方程，有时简称为方程，未知函数是一元函数的微分方程称作常微分方程，未知函数是多元函数的微分方程称作偏微分方程。

An equation that contains parameters, an unknown function, and derivatives (or differentials) of that unknown function is called a differential equation, sometimes simply an equation. A differential equation in which the unknown function is a function of one variable is called an ordinary differential equation (ODE); one in which the unknown function is a multivariate function is called a partial differential equation (PDE).

$$f\left(x, \frac{d^n y}{dx^n},\frac{d^{(n-1)} y}{dx^{(n-1)}},\cdots, \frac{dy}{dx}, y\right)=0$$

SymPy提供了`sympy.dsolve`函数用于求解常微分方程（[ODE](https://docs.sympy.org/dev/modules/solvers/ode.html)），和`sympy.pdsolve`函数用于求解偏微分方程（[PDE](https://docs.sympy.org/dev/modules/solvers/pde.html)）。可选参数hint指明了求解使用的方法。

SymPy provides the `sympy.dsolve` function for solving ordinary differential equations ([ODE](https://docs.sympy.org/dev/modules/solvers/ode.html)), and the `sympy.pdsolve` function for solving partial differential equations ([PDE](https://docs.sympy.org/dev/modules/solvers/pde.html)). The optional hint argument specifies the method used for solving.

<!-- bilingual -->

### 常微分方程 / Ordinary Differential Equations

无摩擦平面上，固定在理想弹簧上质点的运动方程可以用以下一阶齐次线性方程描述：
$$\frac{d^2x}{dt^2}=-kx$$

On a frictionless plane, the equation of motion of a particle attached to an ideal spring can be described by the following first-order homogeneous linear equation:
$$\frac{d^2x}{dt^2}=-kx$$

<!-- bilingual -->

```python
t, k = sympy.symbols("t, k", positive=True)
x = sympy.Function('x')(t)
g = sympy.diff(x, t, 2) + k*x
ans = sympy.dsolve(g, x)
ans
```

对于阻尼系统，其运动方程为：
$$\frac{d^2x}{dt^2}=-kx-c\frac{dx}{dt} $$

For a damped system, its equation of motion is:
$$\frac{d^2x}{dt^2}=-kx-c\frac{dx}{dt} $$

<!-- bilingual -->

```python
t, k, c = sympy.symbols("t, k, c", positive=True)
x = sympy.Function('x')
g = sympy.diff(x(t), t, 2) + c*sympy.diff(x(t), t)  + k*x(t)
ans = sympy.dsolve(g, x(t))
ans
```
