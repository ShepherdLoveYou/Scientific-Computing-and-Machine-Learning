# 分子动力学 / Molecular dynamics

---

分子动力学是一套分子模拟方法，该方法主要是依靠计算机来模拟分子、原子体系的运动，是一种多体模拟方法。通过对分子、原子在一定时间内运动状态的模拟，从而以动态观点考察系统随时间演化的行为。通常，分子、原子的轨迹是通过数值求解牛顿运动方程得到，势能（或其对笛卡尔坐标的一阶偏导数，即力）通常可以由分子间相互作用势能函数、分子力学力场、全始计算给出。对于考虑分子本身的量子效应的体系，往往采用波包近似处理或采用量子力学的费恩曼路径积分表述方式处理。 分子动力学也常常被采用作为研究复杂体系热力学性质的采样方法。在分子体系的不同状态构成的系综中抽取样本，从而计算体系的构型积分，并以构型积分的结果为基础进一步计算体系的热力学量和其他宏观性质。 分子动力学最早在20世纪50年代由物理学家提出，如今广泛应用于物理、化学、生物体系的理论研究中。

Molecular dynamics is a family of molecular simulation methods that rely on computers to simulate the motion of molecular and atomic systems; it is a many-body simulation method. By simulating the motion of molecules and atoms over a period of time, one can examine the evolution of a system over time from a dynamical point of view. Usually, the trajectories of molecules and atoms are obtained by numerically solving Newton's equations of motion, while the potential energy (or its first-order partial derivatives with respect to Cartesian coordinates, i.e. the forces) can be given by intermolecular interaction potential functions, molecular mechanics force fields, or first-principles calculations. For systems in which the quantum effects of the molecules themselves must be considered, the wave packet approximation or the Feynman path integral formulation of quantum mechanics is often used. Molecular dynamics is also often used as a sampling method to study the thermodynamic properties of complex systems. By sampling from an ensemble formed by different states of a molecular system, one computes the configurational integral of the system, and on the basis of the configurational integral further computes the thermodynamic quantities and other macroscopic properties of the system. Molecular dynamics was first proposed by physicists in the 1950s, and is now widely used in theoretical studies of physical, chemical, and biological systems.

## 经典模型 / Classical models

---

玻恩-奥本海默近似（Born-Oppenheimer approximation），又称为绝热近似，将原子核和电子的运动分开处理，因为它们的质量和运动速度有很大的差别。这给理论化学家提供了两种选择，即如何对化学系统进行建模，并确定势能：

The Born-Oppenheimer approximation, also known as the adiabatic approximation, treats the motion of nuclei and electrons separately, because their masses and speeds differ greatly. This gives theoretical chemists two options for how to model a chemical system and determine its potential energy:

* 把原子看作是静止的，用薛定谔方程来模拟电子的运动和相互作用。
* 将电子的运动整合到原子核中，并将这些粒子建模为点电荷。对某一特定原子构型的势能的了解意味着有可能比较不同的构型并确定最可能的构型。

* Treat atoms as stationary and use the Schrödinger equation to simulate the motion and interaction of electrons.
* Integrate the motion of the electrons into the atomic nuclei and model these particles as point charges. Knowing the potential energy of a particular atomic configuration makes it possible to compare different configurations and determine the most likely one.

前者是量子力学计算的基础，如密度函数理论（DFT）方法。在这些方法中，目的是找到薛定谔方程的迭代解。然而，这些方法在计算上非常昂贵，因此现实中仅限于数百或数千个原子。

The former is the foundation of quantum-mechanical calculations, such as density functional theory (DFT) methods. In these methods, the goal is to find iterative solutions of the Schrödinger equation. However, these methods are very expensive computationally, and are therefore limited in practice to hundreds or thousands of atoms.

后者使用的是被称为经典方法的方法。经典方法涉及使用势能模型（有时称为力场）来模拟化学系统。可能利用经典方法的模拟技术的例子包括分子动力学（我们将在本资料中介绍）、蒙特卡洛、朗文动力学等。势能模型使用数学函数来确定原子的特定构型的势能。使用数学函数来模拟粒子的相互作用，其计算成本比量子力学方法低，这意味着可以模拟更大的系统。

The latter uses what are called classical methods. Classical methods involve using potential energy models (sometimes called force fields) to simulate chemical systems. Examples of simulation techniques that may use classical methods include molecular dynamics (which we will introduce in this material), Monte Carlo, Langevin dynamics, etc. Potential energy models use mathematical functions to determine the potential energy of a specific configuration of atoms. Using mathematical functions to model particle interactions is computationally cheaper than quantum-mechanical methods, which means larger systems can be simulated.

### 1. 范德瓦尔斯相互作用 / 1. Van der Waals interactions

一个经典模型的例子就是只存在范德瓦尔斯相互作用的粒子，例如惰性元素气体Ar。我们需要一个势能模型，即一个数学公式，用以描述这种相互作用。众所周知，范德瓦耳斯相互作用包括两种力，第一种是长程吸引的伦敦色散力，第二种是短程泡利排斥力。

An example of a classical model is a particle in which only van der Waals interactions exist, for example the noble gas Ar. We need a potential energy model, i.e. a mathematical formula, to describe this interaction. It is well known that the van der Waals interaction consists of two kinds of forces: first, the long-range attractive London dispersion force, and second, the short-range Pauli repulsion force.

用于模拟范德瓦尔斯相互作用的常用数学函数是Lennard-Jones势。这个模型考虑了长程吸引力：

A commonly used mathematical function for modeling van der Waals interactions is the Lennard-Jones potential. This model accounts for the long-range attraction:

$$E_{\text{attractive}}(r) = -4\varepsilon\Big(\dfrac{\sigma}{r}\Big)^6 $$

其中σ是两个粒子之间势能为零的距离，-ε是平衡分离时的势能，而r是两个原子之间的距离。

where σ is the distance at which the potential energy between the two particles is zero, -ε is the potential energy at the equilibrium separation, and r is the distance between the two atoms.

同时，这个模型也考虑了泡利不相容原理导致的短程排斥力：

The model also accounts for the short-range repulsion due to the Pauli exclusion principle:

$$E_{\text{repulsive}}(r) = 4\varepsilon\Big(\dfrac{\sigma}{r}\Big)^{12} $$

相应的，相互作用力的数学函数为：

Correspondingly, the mathematical function for the interaction force is:

$$F = \dfrac{-\partial E(r)}{\partial r} = 48\varepsilon\dfrac{1}{r}\Big(\dfrac{\sigma}{r}\Big)^{12} - 24\varepsilon\dfrac{1}{r}\Big(\dfrac{\sigma}{r}\Big)^6 $$

<!-- bilingual -->

```python
import numpy as np

def LJ_potential(r, epsilon, sigma):
        """
        :param float r: Distance between particles (Å)
        :param float epsilon: Potential energy at the equilibrium distance (eV)
        :param float sigma: Distance at which the potential energy is zero (Å)
        :returns: Lennard-Jones potential (eV)
        """
        rm_6 = np.power(sigma/r, 6)
        return 4 * epsilon * rm_6**2 - 4 * epsilon * rm_6

def LJ_force(r, epsilon, sigma):
        """
        :param float r: Distance between particles (Å)
        :param float epsilon: Potential energy at the equilibrium distance (eV)
        :param float sigma: Distance at which the potential energy is zero (Å)
        :returns: Lennard-Jones force (eV/Å)
        """
        rm_6 = np.power(sigma/r, 6)
        return 48 * epsilon * rm_6**2 / r - 24 * epsilon * rm_6 / r
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

r = np.linspace(2, 8, 100)
epsilon = 0.013
sigma = 3.4
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(r, LJ_potential(r, epsilon, sigma), 'r', alpha=0.8)
ax.set_ylim(-0.02, 0.04)
ax.set_xlabel(r'$r$ (Å)', fontsize=14)
ax.set_ylabel(r'$E$ (eV)', fontsize=14, color="red")
ax.axhline(0, ls="--")
ax.axvline(sigma, ls="--")

ax2 = ax.twinx()
ax2.plot(r, LJ_force(r, epsilon, sigma), 'b', alpha=0.8)
ax2.set_ylim(-0.02, 0.04)
ax2.set_ylabel(r'$F$ (eV/Å)', fontsize=14, color="blue")
```

一般来说，Lennard-Jonesz之类的势能模型只可以描述自相互作用，即一个特定原子与另一个同类型原子之间的范德瓦尔斯相互作用。不同类型的原子之间如何相互作用是通过使用混合规则来实现的。例如，最常见的Lorentz-Berthelot规则：

In general, potential energy models such as Lennard-Jones can only describe self-interactions, i.e. the van der Waals interaction between a particular atom and another atom of the same type. The interaction between different types of atoms is handled via combining rules. For example, the most common Lorentz-Berthelot rules:

$$\sigma_{ij} = \dfrac{\sigma_{ii} + \sigma_{jj}}{2} \;\;\;\text{and}\;\;\; \varepsilon_{ij} = \sqrt{\varepsilon_{ii}\varepsilon_{jj}}$$

### 2. 静电相互作用 / 2. Electrostatic interactions

除了范德瓦尔斯相互作用之外，另一类重要的非共价相互作用是带电粒子之间的静电相互作用，使用库仑定律描述：

In addition to van der Waals interactions, another important class of non-covalent interactions is the electrostatic interaction between charged particles, described by Coulomb's law:

$$E_{\text{Coulomb}}(r) = \frac{1}{4\pi\varepsilon_0}\frac{q_iq_je^2}{r}$$

其中，$q_i$和$q_j$是粒子上的电子电荷，e是电子的电荷，$\varepsilon_0$是真空的介电率，r是两个点电荷之间的距离。

where $q_i$ and $q_j$ are the electron charges on the particles, e is the electron charge, $\varepsilon_0$ is the vacuum permittivity, and r is the distance between the two point charges.

相应的，库仑力为：

Correspondingly, the Coulomb force is:

$$F = \dfrac{-\partial E(r)}{\partial r} = \frac{1}{4\pi\varepsilon_0}\frac{q_iq_je^2}{r^2} $$

<!-- bilingual -->

```python
import numpy as np
from scipy.constants import elementary_charge, epsilon_0

def Coulomb_potential(qi, qj, r):
        """
        :param int qi: Electronic charge on particle i
        :param int qj: Electronic charge on particle j
        :param float r: Distance between particles (Å)
        :returns: Coulomb potential (eV)
        """
        energy_joules = qi*qj*elementary_charge**2 / (4*np.pi*epsilon_0*r*1e-10)
        return energy_joules / elementary_charge

def Coulomb_force(qi, qj, r):
        """
        :param int qi: Electronic charge on particle i
        :param int qj: Electronic charge on particle j
        :param float r: Distance between particles (Å)
        :returns: Force of the Coulombic interaction (eV/Å)
        """
        energy_newton = qi*qj*elementary_charge**2 / (4*np.pi*epsilon_0*r**2*1e-20)
        return energy_newton / elementary_charge * 1e-10
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

r = np.linspace(2, 8, 100)
qi = 1
qj = -1
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(r, Coulomb_potential(qi, qj, r), 'r', alpha=0.8)
ax.set_ylim(-6, 0)
ax.set_xlabel(r'$r$ (Å)', fontsize=14)
ax.set_ylabel(r'$E$ (eV)', fontsize=14, color="red")

ax2 = ax.twinx()
ax2.plot(r, Coulomb_force(qi, qj, r), 'b', alpha=0.8)
ax2.set_ylim(-6, 0)
ax2.set_ylabel(r'$F$ (eV/Å)', fontsize=14, color="blue")
```

### 3. 化学键 / 3. Chemical bonds

借助范德瓦尔斯相互作用和静电相互作用对原子进行模拟对于研究非共价材料非常有用。然而，为了真正探究一个系统的化学特性，我们还必须能够对化学键的相互作用进行数学建模。

Using van der Waals and electrostatic interactions to simulate atoms is very useful for studying non-covalent materials. However, in order to truly probe the chemical properties of a system, we must also be able to mathematically model the interactions of chemical bonds.

一个系统的化学键合的相互作用的势能通常是由键长、键角、二面角（扭转角）和其他高阶过程组成：

The potential energy of the bonded interactions of a system is usually composed of bond length, bond angle, dihedral (torsion) angle, and other higher-order processes:

$$E_{\text{bonded}} = E_{\text{bond}}(b) + E_{\text{angle}}(\theta) + E_{\text{dihedral}}(\phi) + \ldots$$

这里我们只介绍前两项，键长、键角对势能的贡献。

Here we only introduce the first two terms, the contributions of bond length and bond angle to the potential energy.

<!-- bilingual -->

#### 键长 / Bond length

键长的影响通常可以用简谐势能函数来模拟：

The effect of bond length can typically be modeled by a harmonic potential energy function:

$$E_{\text{bond}}(\mathbf{r}) = \frac{K_b}{2}(r - b_0)^2$$

其中，$b_0$是平衡键长，$K_b$是该键的弹性系数。这些参数在使用前必须在借助实验或者第一性原理计算测定。

where $b_0$ is the equilibrium bond length and $K_b$ is the stiffness constant of the bond. These parameters must be determined via experiment or first-principles calculations before use.

下面是这个函数的一个例子，其中使用$H_2$分子中H-H键的参数。

The following is an example of this function, using the parameters for the H-H bond in the $H_2$ molecule.

<!-- bilingual -->

```python
%matplotlib inline
import matplotlib.pyplot as plt

def bond_length(kb, b0, r):
    """
    :param float kb: Bond force constant (eV/Å^2)
    :param float b0: Equilibrium bond length (Å)
    :param float b: Bond length (Å)
    :returns: Energy of the bonded interaction (eV)
    """
    return kb / 2 * (r - b0) ** 2

r = np.linspace(0.7, 0.8, 100)
plt.plot(r, bond_length(25, 0.74, r))
plt.xlabel(r'$r$ (Å)')
plt.ylabel(r'$E$ (eV)')
```

如果键长偏离了平衡值，增大或者减小都会导致势能升高。

If the bond length deviates from the equilibrium value, both increasing and decreasing will cause the potential energy to rise.

<!-- bilingual -->

#### 键角 / Bond angle

两个连接同一个原子的化学键之间的角度与势能之间的关系也可以使用一个简谐函数描述：

The relationship between the angle between two chemical bonds sharing the same atom and the potential energy can also be described by a harmonic function:

$$E_{\text{angle}}(\theta) = \frac{K_\theta}{2}(\theta - \theta_0)^2$$

其中，$\theta$是测量的角度，$K_\theta$是该角度的力常数，$\theta_0$是平衡角度大小。这个函数与上面显示的键长-势能曲线具有相同的形状。

where $\theta$ is the measured angle, $K_\theta$ is the force constant for this angle, and $\theta_0$ is the equilibrium angle. This function has the same shape as the bond-length-potential-energy curve shown above.

下面以$H_2O$分子举例，其中参数为H-O键之间的键角。

The following uses the $H_2O$ molecule as an example, where the parameters are for the angle between H-O bonds.

<!-- bilingual -->

```python
%matplotlib inline
import matplotlib.pyplot as plt

def bond_angle(k_theta, theta_0, theta):
    """
    :param float kb: Bond force constant (eV/Arcdegree^2)
    :param float b0: Equilibrium bond angle (Arcdegree)
    :param float b: Bond angle (Arcdegree)
    :returns: Energy of the bonded interaction (eV)
    """
    return k_theta / 2 * (theta - theta_0) ** 2

theta = np.linspace(90, 120, 100)
plt.plot(theta, bond_angle(4e-4, 104.45, theta))
plt.xlabel(r'$r$ (Å)')
plt.ylabel(r'$E$ (eV)')
```

* 时间 time = femtoseconds
* 能量 energy = eV

* mass = a.m.u or grams/mole
* distance = Angstroms
* time = femtoseconds
* energy = eV
* force = eV/Angstrom
* temperature = Kelvin
* charge = multiple of electron charge (1.0 is a proton)

在应用物理公式时，要遵循相应的转换规则。

When applying physical formulas, the corresponding conversion rules must be followed.

例如，计算数据中$a$的单位使用$[\mathbf{Angstrom/fs^{2}}]$，但是$\frac{\mathbf{f}}{m}$的单位为$[\mathbf{\frac{eV/Angstrom}{amu}}]$，则牛顿第二定律计算的加速度数值应当做如下变换：

For example, the unit of $a$ in the data is $[\mathbf{Angstrom/fs^{2}}]$, but the unit of $\frac{\mathbf{f}}{m}$ is $[\mathbf{\frac{eV/Angstrom}{amu}}]$; the acceleration computed from Newton's second law should be converted as follows:

$$[\mathbf{\frac{eV/Angstrom}{amu}}] = {Avogadro \times ElementaryCharge \times 10^{-7}} [\mathbf{Angstrom/fs^{2}}]$$

其中 Avogadro 为阿伏伽德罗常数，ElementaryCharge 为质子电量。

where Avogadro is Avogadro's constant and ElementaryCharge is the proton charge.

## 分子动力学模拟 / Molecular dynamics simulation

我们使用分子动力学方法和Lennard-Jones势，对惰性单原子气体Ar的二维热运动进行模拟。

We use the molecular dynamics method and the Lennard-Jones potential to simulate the two-dimensional thermal motion of the noble monatomic gas Ar.

### 1. 力和加速度 / 1. Forces and accelerations

我们研究的粒子的运动是纯经典的，因此我们可以应用经典力学来研究其动态行为。牛顿第二运动定律：

The motion of the particles we study is purely classical, so we can apply classical mechanics to study their dynamic behavior. Newton's second law of motion:

$$\mathbf{f} = m\mathbf{a}$$

其中$\mathbf{f}$是质量为$m$的原子上受到的合外力，其加速度为$\mathbf{a}$。

where $\mathbf{f}$ is the net external force on an atom of mass $m$, and its acceleration is $\mathbf{a}$.

任意两个粒子之间的力相互作用力$\mathbf{f}$，可以从相互作用能量$E(r)$中找到。

The interaction force $\mathbf{f}$ between any two particles can be found from the interaction energy $E(r)$:

$$f = \dfrac{-\partial E(r)}{\partial r}$$

这就是说，力是能量相对于粒子位置的一阶导数的负数。我们将使用前面创建的函数，从Lennard-Jones势中计算出力。

That is, the force is the negative of the first derivative of the energy with respect to the particle position. We will use the functions created earlier to compute the force from the Lennard-Jones potential.

<!-- bilingual -->

#### Lennard-Jones truncated & shifted (LJTS) potential

在实际运算中，出于计算效率的考虑，可以忽略Lennard-Jones的长程作用，将其在特定距离上进行截断，从而得到LJTS势。最常用的截断距离是平衡距离的2.5倍。

In practical computation, for the sake of computational efficiency, the long-range part of the Lennard-Jones interaction can be ignored and truncated at a particular distance, yielding the LJTS potential. The most commonly used truncation distance is 2.5 times the equilibrium distance.

<!-- bilingual -->

```python
import numpy as np

def LJ_shifted_potential(r, epsilon, sigma):
        """
        :param float r: Distance between particles (Å)
        :param float epsilon: Potential energy at the equilibrium distance (eV)
        :param float sigma: Distance at which the potential energy is zero (Å)
        :returns: Lennard-Jones potential (eV)
        """
        r_cutoff = 2.5 * sigma
        r_cutoffm_6 = np.power(sigma/r_cutoff, 6)
        potential_shift = 4 * epsilon * r_cutoffm_6**2 - 4 * epsilon * r_cutoffm_6
        rm_6 = np.power(sigma/r, 6)
        LJ_shifted_potential = 4 * epsilon * rm_6**2 - 4 * epsilon * rm_6
        return LJ_shifted_potential - potential_shift
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

r = np.linspace(2, 10, 100)
epsilon = 0.013
sigma = 3.4
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(r, LJ_potential(r, epsilon, sigma), 'r', alpha=0.8)
ax.plot(r, LJ_shifted_potential(r, epsilon, sigma), 'b', alpha=0.8)
ax.set_ylim(-0.015, 0.005)
ax.set_xlabel(r'$r$ (Å)', fontsize=14)
ax.set_ylabel(r'$E$ (eV)', fontsize=14, color="red")
ax.axhline(0, ls="--")
ax.axvline(sigma, ls="--")
ax.axvline(2.5 * sigma, ls="--")
```

### 2. 数值积分 / 2. Numerical integration

牛顿运动方程是一个二阶微分方程，虽然我们无法使用解析方法处理描述所有粒子运动的微分方程组，但是我们可以借助因变量离散化思想，使用数值方法求解其轨迹方程。

Newton's equations of motion form a second-order differential equation. Although we cannot analytically solve the ODE system describing the motion of all particles, we can discretize the dependent variables and use numerical methods to solve their trajectory equations.

$$\mathbf{x}_i(t + \Delta t) = \mathbf{x}_i(t) + \mathbf{v}_i(t)\Delta t + \dfrac{1}{2} \mathbf{a}_i(t)\Delta t^2$$
$$\mathbf{v}_i(t + \Delta t) = \mathbf{v}_i(t) + \dfrac{1}{2}\big[\mathbf{a}_i(t) + \mathbf{a}_i(t+\Delta t)\big]\Delta t$$

其中，$\Delta t$是时间步长（时间的增量），$\mathbf{x}_i$是粒子位置，$\mathbf{v}_i$是速度，$\mathbf{a}_i$是加速度。这对方程被称为Velocity-Verlet算法，其基本流程为：
1. 计算粒子上的力（得到加速度）
2. 计算某个时间段后粒子的位置
3. 在新位置上计算新的力和新的加速度
4. 根据旧位置和新位置的平均加速度，确定粒子的新速度
5. 重复以上步骤

where $\Delta t$ is the time step (the increment of time), $\mathbf{x}_i$ is the particle position, $\mathbf{v}_i$ is the velocity, and $\mathbf{a}_i$ is the acceleration. This pair of equations is called the Velocity-Verlet algorithm, and its basic procedure is:
1. Compute the force on each particle (to obtain the acceleration)
2. Compute the position of each particle after a time step
3. Compute the new forces and accelerations at the new positions
4. Based on the average of the old and new accelerations, determine the new velocity of the particle
5. Repeat the above steps

<!-- bilingual -->

```python
import numpy as np
from scipy.spatial.distance import pdist, squareform
from scipy.constants import Avogadro, elementary_charge

class MDsimulation:

    def __init__(self, pos, vel, box_size, m=39.948, epsilon=0.0103, sigma=3.4, pbc=True):
        """
        Initialize the simulation with identical particles of mass m.
        The intermolecular interactions are described using Lennard-Jones potential.
        The n x 2 state arrays pos and vel hold the n particles'
        positions in their rows as (x_i, y_i) and (vx_i, vy_i).
        """
        self.pos = np.array(pos, dtype=float)  # Å
        self.vel = np.array(vel, dtype=float)  # Å/fs
        self.box_size = box_size  # Å
        self.n = self.pos.shape[0]
        self.m = m / Avogadro / elementary_charge * 1e7  # amu to eVfs^2/Å^2
        self.epsilon = epsilon  # eV
        self.sigma = sigma  # Å
        self.r_cutoff = 2.5 * sigma  # Å
        self.pbc = pbc
        self.force = self.update_force(pos)  # eV/Å, Å
        self.t = 0  # femtosecond

    def LJ_shifted_potential(self, r):
        """
        Lennard-Jones truncated & shifted (LJTS) potential
        :param float r: Distance between particles (Å)
        :param float epsilon: Potential energy at the equilibrium distance (eV)
        :param float sigma: Distance at which the potential energy is zero (Å)
        :returns: Lennard-Jones potential (eV)
        """
        r_cutoffm_6 = np.power(self.sigma/self.r_cutoff, 6)
        potential_shift = 4 * self.epsilon * r_cutoffm_6**2 - 4 * self.epsilon * r_cutoffm_6
        rm_6 = np.power(self.sigma/r, 6)
        LJ_potential = 4 * self.epsilon * rm_6**2 - 4 * self.epsilon * rm_6
        return LJ_potential - potential_shift

    def LJ_shifted_force(self, r):
        """
        Calculate force (eV/Å) using Lennard-Jones truncated & shifted (LJTS) potential
        """
        rm_6 = np.power(self.sigma/r, 6)
        LJ_force = 48 * self.epsilon * rm_6**2 / r - 24 * self.epsilon * rm_6 / r
        r_cutoffm_6 = np.power(self.sigma/self.r_cutoff, 6)
        LJ_shifted_force = 48 * self.epsilon * r_cutoffm_6**2 / r - 24 * self.epsilon * r_cutoffm_6 / r
        return LJ_force - LJ_shifted_force

    def update_force(self, pos):
        """
        Update forces
        """
        force = np.zeros_like(pos)
        if self.pbc is True:
            pos_left = pos[pos[:,0] < self.r_cutoff] + [self.box_size[0],0]
            pos_right = pos[pos[:,0] > (self.box_size[0] - self.r_cutoff)] - [self.box_size[0],0]
            pos_extended = np.concatenate((pos, pos_left, pos_right))
            pos_top =  pos[pos[:,1] < self.r_cutoff] + [0, self.box_size[1]]
            pos_bottom =  pos[pos[:,1] > (self.box_size[1] - self.r_cutoff)] - [0, self.box_size[1]]
            self.pos_extended = np.concatenate((pos_extended, pos_top, pos_bottom))
        else:
            self.pos_extended = pos
        self.dist = squareform(pdist(self.pos_extended))
        iarr, jarr = np.where(self.dist < self.r_cutoff)
        for i, j in zip(iarr, jarr):
            if i == j: continue
            if i >= self.n: continue
            r = self.dist[i, j]
            force[i] += self.LJ_shifted_force(r) * (self.pos_extended[i] - self.pos_extended[j]) / r
        return force

    def advance(self, dt):
        """
        Advance the simulation by dt femtosecond.
        """
        self.t += dt
        # Update the particles' positions and velocities using Velocity-Verlet algorithm
        a_0 = self.force / self.m
        self.pos += self.vel * dt + 0.5 * a_0 * dt**2
        self.force = self.update_force(self.pos)
        a_1 = self.force / self.m
        self.vel += 0.5 * (a_0 + a_1) * dt
        if self.pbc is True:
            self.pos = self.pos % self.box_size
        else:
            # Bounce the particles off the walls by reflecting their velocity vectors
            hit_left_wall = self.pos[:, 0] < 0
            hit_right_wall = self.pos[:, 0] > self.box_size[0]
            hit_bottom_wall = self.pos[:, 1] < 0
            hit_top_wall = self.pos[:, 1] > self.box_size[1]
            self.vel[hit_left_wall | hit_right_wall, 0] *= -1
            self.vel[hit_bottom_wall | hit_top_wall, 1] *= -1

    def get_potential_energy(self):
        """
        :param float vel: Positions of particles (Å)
        :returns: Total potential_energy (eV)
        """
        potential_energy = 0.0
        iarr, jarr = np.where(self.dist > 0.1)
        for i, j in zip(iarr, jarr):
            if i >= j or i >= self.n: continue
            r = self.dist[i, j]
            potential_energy += self.LJ_shifted_potential(r)
        return potential_energy

    def get_kinetic_energy(self):
        """Returns the total kinetic_energy (eV) """
        return np.sum(0.5 * self.m * np.sum((self.vel)**2, axis=1))

    def get_temperature(self):
        """Returns the temperature (K)"""
        mean_kinetic_energy = self.get_kinetic_energy() / self.n
        return mean_kinetic_energy * elementary_charge / Boltzmann

    def get_pressure(self):
        """Returns the pressure (eV/Å^2)"""
        mean_kinetic_energy = self.get_kinetic_energy() 
        volume = np.prod(self.box_size)
        return mean_kinetic_energy / volume
```

上述过程是单步数值积分，Velocity-Verlet算法是我们使用的积分器。

The above is a single-step numerical integration, and the Velocity-Verlet algorithm is the integrator we use.

这个算法对于两个以上的粒子是高度非线性的，其只对非常小的$\Delta t$有效，如果使用较大的时间步长，计算出的加速度将不准确，因为原子所受外力在该时间间隔内可能发生很大变化。

This algorithm is highly non-linear for more than two particles, and is only valid for very small $\Delta t$. If a larger time step is used, the computed acceleration will be inaccurate, because the external force on an atom can change significantly within that time interval.

时间步长的通常选择是$10^{-15}$秒（飞秒）。因此，为了模拟一纳秒的 "实时 "分子动力学，必须使用上述算法进行一百万次迭代。这对于大型的系统（例如生物分子）来说可能是非常缓慢的。

The time step is typically chosen as $10^{-15}$ seconds (femtoseconds). Therefore, to simulate one nanosecond of "real-time" molecular dynamics, the above algorithm must be iterated a million times. This can be very slow for large systems (e.g. biomolecules).

### 3. 模型初始化 / 3. Model initialization

在运行分子动力学模拟之前，我们还需要设置系统的初始状态：$t=0$时粒子的位置和速度。

Before running a molecular dynamics simulation, we also need to set the initial state of the system: the positions and velocities of the particles at $t=0$.

粒子的位置可以从结构数据库（例如蛋白质结构数据库或者晶体学数据库）获取，或者基于对系统的认识自行构建（例如，具有面心立方结构的CaF2的表面模型）。这个初始结构的重要性怎么强调都不为过。如果初始结构不能代表平衡结构，可能需要很长时间才能得到平衡结构，或者永远无法获得合理结构。

Particle positions can be obtained from structural databases (e.g. the Protein Data Bank or crystallographic databases), or constructed by oneself based on knowledge of the system (for example, a surface model of face-centered-cubic CaF2). The importance of this initial structure cannot be overstated. If the initial structure is not representative of the equilibrium structure, it may take a very long time to reach the equilibrium structure, or a reasonable structure may never be obtained.

对粒子速度进行初始化的过程更为普通。因为系统的总动能$E_K$取决于模拟的温度$T$。

Initializing the particle velocities is a more routine process, because the total kinetic energy $E_K$ of the system depends on the simulation temperature $T$.

$$E_K = \sum_{i=1}^N \dfrac{m_i|v_i|^2}{2} = \dfrac{3}{2}Nk_BT$$

其中，$m_i$是粒子i的质量，N是粒子的数量，$k_B$是玻尔兹曼常数。

where $m_i$ is the mass of particle i, N is the number of particles, and $k_B$ is the Boltzmann constant.

对于我们将要模拟的二维系统，粒子总动能和温度的关系为：

For the two-dimensional system we will simulate, the relation between the total kinetic energy of the particles and the temperature is:

$$E_K = Nk_BT$$

<!-- bilingual -->

```python
from scipy.constants import Boltzmann, Avogadro

def initial_velocity(T, mass):
    """
    :param float T: Temprature (K)
    :param float T: Mass of particle (amu)
    :returns: root mean square velocity of particles (Å/fs)
    """
    mass_Kg = mass / 1e3 / Avogadro
    knetic_energy_joules = Boltzmann * T
    velocity = np.sqrt(2*knetic_energy_joules/mass_Kg)  # m/s
    return velocity / 1e5  # convert to Å/fs

amu_Ar = 39.948
epsilon = 0.0103
sigma = 3.4

# RMS Veocity at room temprature
rms_velocity = initial_velocity(300, amu_Ar)
print('The root mean square velocity of Ar at 300K is %.4f Å/fs' %(rms_velocity))
```

我们将使用如下参数对模型进行初始化：
1. 系统的尺寸为10 nm × 10 nm
2. 粒子的初始位置为随机值，最小间距为3 Å
2. 系统的初始温度为300 K，对应于粒子的初始速率为0.0035 Å/fs
3. 粒子的初始运动方向为随机值
4. 使用NVE系综（孤立系统），即粒子数、系统尺寸和系统总能量保持不变

We will initialize the model with the following parameters:
1. The system size is 10 nm × 10 nm
2. Initial particle positions are random, with a minimum spacing of 3 Å
2. The initial system temperature is 300 K, corresponding to an initial particle speed of 0.0035 Å/fs
3. The initial direction of motion of each particle is random
4. Use the NVE ensemble (isolated system), i.e. the number of particles, system size, and total system energy are kept constant

```python
n = 64
box_size = np.array([100, 100])
pos = np.random.random((n, 2)) * (box_size - sigma) + sigma/2 
while np.min(pdist(pos)) < 3:
    pos = np.random.random((n, 2)) * (box_size - sigma) + sigma/2

vel_dir = np.random.random((n, 2))-0.5
vel = rms_velocity * vel_dir / np.atleast_2d(np.linalg.norm(vel_dir, axis=1)).T
```

## 系综 / Ensembles

---

在统计物理中，系综（ensemble）代表一定条件下一个体系的大量可能状态的集合。也就是说，系综是系统状态的一个概率分布。具有相同宏观性质（例如温度）的体系，其微观状态（例如粒子的位置和速度）仍然可以大不相同。实际上，对于一个宏观体系，所有可能的微观状态数是天文数字。

In statistical physics, an ensemble represents the collection of a large number of possible states of a system under certain conditions. In other words, an ensemble is a probability distribution over the states of the system. Systems with the same macroscopic properties (e.g. temperature) can still have very different microscopic states (e.g. particle positions and velocities). In fact, for a macroscopic system, the number of all possible microscopic states is astronomical.

统计物理的一个原理（各态历经原理）是：对于一个处于平衡的体系，物理量的时间平均，等于对对应系综里所有体系进行平均的结果。体系的平衡态的物理性质可以对不同的微观状态求和来得到。

One principle of statistical physics (the ergodic principle) states: for a system in equilibrium, the time average of a physical quantity equals the result of averaging over all systems in the corresponding ensemble. The physical properties of the equilibrium state of a system can be obtained by summing over different microstates.

前面介绍的分子动力学模拟示例使用了NVE系综或微正则系综（microcanonical ensemble），其中粒子数（N）、系统体积（V）和系统能量（E）都保持不变。

The molecular dynamics simulation example introduced earlier uses the NVE or microcanonical ensemble, in which the number of particles (N), system volume (V), and system energy (E) are all kept constant.

常用的系综还有：
* NVT系综或正则系综（canonical  ensemble），其中粒子数（N）、系统体积（V）和系统温度（T）都保持不变，系综内各体系可以和外界交换能量。
* NPT系综或等温等压系综（isothermal-isobaric ensemble）：正则系综的推广，各体系可以和外界环境交换能量和体积，但系综内各个体系有相同的温度和压强。

Other commonly used ensembles include:
* NVT or canonical ensemble, in which the number of particles (N), system volume (V), and system temperature (T) are kept constant, and the individual systems in the ensemble can exchange energy with the environment.
* NPT or isothermal-isobaric ensemble: a generalization of the canonical ensemble, where individual systems can exchange energy and volume with the environment, but all systems within the ensemble have the same temperature and pressure.

### 恒温器 / Thermostat

对于这些系综，有必要确定一种方法来调节系统的温度。最简单（不一定最合理）的方法，是速度缩放，使整个系统的动能更准确地符合所需温度的要求。系统的瞬时温度$T_{\text{inst}}$被定义为：

For these ensembles, it is necessary to determine a way to regulate the system's temperature. The simplest (not necessarily the most reasonable) method is velocity scaling, which makes the kinetic energy of the whole system match the requirement of the target temperature more accurately. The instantaneous temperature $T_{\text{inst}}$ of the system is defined as:

$$T_{\text{inst}} = \frac{\sum^N_{i=0}{m_i v_i^2}}{2Nk_B}$$

其中，N是粒子的数量，$m_i$是粒子i的质量，$v_i$是粒子i的速度，$k_B$是玻尔兹曼常数。

where N is the number of particles, $m_i$ is the mass of particle i, $v_i$ is the velocity of particle i, and $k_B$ is the Boltzmann constant.

这意味着，粒子的速度可以通过以下关系来重新分配:

This means that the particle velocities can be redistributed according to the following relation:

$$v_i = v_i \sqrt{\dfrac{T_{\text{target}}}{\bar{T}}}$$

其中$T_{\text{target}}$是恒温器的目标温度，$\bar{T}$是平均模拟温度。

where $T_{\text{target}}$ is the target temperature of the thermostat, and $\bar{T}$ is the average simulation temperature.

还有其他各种恒温方法，如Anderson、Nosé-Hoover或Berendsen方法。

There are a variety of other thermostat methods, such as Anderson, Nosé-Hoover, or Berendsen.

为了实现NPT系综，除了恒温器之外，还必须使用限流器。限流器用于控制系统的体积变化，从而使压力在整个模拟过程中保持不变。

To implement the NPT ensemble, in addition to a thermostat, a barostat must also be used. The barostat is used to control the system's volume change, so that the pressure remains constant throughout the simulation.

<!-- bilingual -->

```python
class MD_NVT(MDsimulation):

    def thermostat(self, target_T, Q=5):
        '''
        Thermostat
        '''
        kinetic_energy = self.get_kinetic_energy() / self.n
        target_kinetic_energy = Boltzmann * target_T / elementary_charge
        return -(kinetic_energy - target_kinetic_energy) / Q * self.vel

    def advance(self, dt, target_T=None):
        '''
        Advance the simulation by dt femtosecond.
        '''
        self.t += dt
        # Update the particles' positions and velocities using Velocity-Verlet algorithm
        a_0 = self.force / self.m
        if target_T is not None:
            a_0 += self.thermostat(target_T)
        self.pos += self.vel * dt + 0.5 * a_0 * dt**2
        self.force = self.update_force(self.pos)
        a_1 = self.force / self.m
        self.vel += 0.5 * (a_0 + a_1) * dt
        if self.pbc is True:
            self.pos = self.pos % self.box_size
        else:
            # Bounce the particles off the walls by reflecting their velocity vectors
            hit_left_wall = self.pos[:, 0] < 0
            hit_right_wall = self.pos[:, 0] > self.box_size[0]
            hit_bottom_wall = self.pos[:, 1] < 0
            hit_top_wall = self.pos[:, 1] > self.box_size[1]
            self.vel[hit_left_wall | hit_right_wall, 0] *= -1
            self.vel[hit_bottom_wall | hit_top_wall, 1] *= -1
```

```python
%matplotlib widget
from matplotlib import pyplot as plt
from matplotlib.animation import FuncAnimation

fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, aspect='equal', autoscale_on=False)
ax.set_xlim(0, sim_NVT.box_size[0])
ax.set_ylim(0, sim_NVT.box_size[1])
ax.set_xticks([])
ax.set_yticks([])

vel_norm = np.linalg.norm(sim_NVT.vel, axis=1)
colors = [plt.cm.rainbow(p) for p in vel_norm/vel_norm.max()]
particles = ax.scatter(sim_NVT.pos[:,0], sim_NVT.pos[:,1], c=colors, s=100)
label = ax.text(0.05 * sim_NVT.box_size[0], 0.05 * sim_NVT.box_size[1], '$time$ = {:.2f} ps and $temperature$ = {:.1f} K'.format(0, 0), size=12)

target_T = 300
dt = 10
def animate_NVT(i):
    for nsteps in range(50):  # plot every 0.5 ps
        sim_NVT.advance(dt, target_T)
    particles.set_offsets(sim_NVT.pos)
    vel_norm = np.linalg.norm(sim_NVT.vel, axis=1)
    colors = [plt.cm.rainbow(p) for p in vel_norm/vel_norm.max()]
    particles.set_color(colors)
    label.set_text('$time$ = {:.1f} ps and $temperature$ = {:.1f} K'.format(sim_NVT.t/1000, sim_NVT.get_temperature()))
    return particles, label

anim_NVT = FuncAnimation(fig, animate_NVT, frames=200, interval=100, repeat=False, blit=False)

#anim.save('/tmp/Ar_gas_NVT.gif', writer='pillow', fps=10)
```
