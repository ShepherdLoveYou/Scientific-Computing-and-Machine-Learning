## 异常处理 / Exception Handling

try 语句按照如下方式工作：
>1. 首先，执行 try 子句（在关键字 try 和关键字 except 之间的语句）。
>2. 如果没有异常发生，忽略 except 子句，try 子句执行后结束。
>3. 如果在执行 try 子句的过程中发生了异常，那么 try 子句余下的部分将被忽略。如果异常的类型和 except 之后的名称相符，那么对应的 except 子句将被执行。
>4. 如果一个异常没有与任何的 except 匹配，那么这个异常将会传递给上层的 try 中。

The try statement works as follows:
>1. First, the try clause (the statements between the keywords try and except) is executed.
>2. If no exception occurs, the except clause is skipped and execution of the try clause ends.
>3. If an exception occurs during execution of the try clause, the rest of the clause is skipped. If the exception's type matches the name after an except, the corresponding except clause is executed.
>4. If an exception is not matched by any except clause, it is propagated to the enclosing try.

<!-- bilingual -->

```python
while True:
    try:
        x = int(input("请输入一个数字: "))
        break
    except ValueError:
        print("您输入的不是数字，请再次尝试输入！")
```

## 迭代器 / Iterators

>1. 迭代器是一个可以记住遍历的位置的对象。
>2. 迭代器对象从集合的第一个元素开始访问，直到所有的元素被访问完结束。迭代器只能往前不会后退。
>3. 字符串，列表或元组对象都可用于创建迭代器。

>1. An iterator is an object that remembers the position it is traversing.
>2. An iterator object starts from the first element of the collection and accesses elements until all of them have been accessed. Iterators can only move forward — they cannot go back.
>3. Strings, lists, and tuples can all be used to create iterators.

<!-- bilingual -->

```python
list=[1,2]
it = iter(list)    # 创建迭代器对象
print(type(it))
print (next(it), next(it))   # 依次输出迭代器的下一个元素
# 迭代器对象可以使用常规for语句进行遍历
it = iter(list)   # 再次创建迭代器对象
for x in it:
    print (x, end=" ")
```

## 生成器 / Generators

>1. 使用了 yield 的函数被称为生成器（generator）。生成器是一个返回迭代器的函数。
>2. 在调用生成器运行的过程中，每次遇到 yield 时函数会暂停并保存当前所有的运行信息，返回 yield 的值, 并在下一次执行 next() 方法时从当前位置继续运行。
>3. 调用一个生成器函数，返回的是一个迭代器对象。

>1. A function that uses yield is called a generator. A generator is a function that returns an iterator.
>2. While a generator is being executed, each time it reaches yield the function is paused, its current execution state is saved, the value of yield is returned, and execution resumes from the current position the next time next() is called.
>3. Calling a generator function returns an iterator object.

<!-- bilingual -->

```python
def fibonacci(n): # 生成器函数 - 斐波那契
    a, b, counter = 0, 1, 0
    while True:
        if (counter > n): 
            return
        yield a
        a, b = b, a + b
        counter += 1
f = fibonacci(5) # f是一个迭代器，由生成器返回生成
print(type(f))
for x in f:
    print(x, end=" ")
```

## 匿名函数 / Anonymous Functions

>1. lambda只是一个表达式，函数体比def简单得多。
>2. lambda的主体是一个表达式，而不是一个代码块，所以不能写太多的逻辑进去。
>3. lambda函数拥有自己的命名空间，且不能访问自有参数列表之外或全局命名空间里的参数。
>4. lambda定义的函数的返回值就是表达式的返回值，不需要return语句块。
>5. lambda表达式的主要应用场景就是赋值给变量，作为参数传入其他函数

>1. A lambda is just a single expression — its body is much simpler than that of def.
>2. The body of a lambda is a single expression rather than a code block, so it cannot contain too much logic.
>3. A lambda function has its own namespace and cannot access parameters outside its own parameter list or in the global namespace.
>4. The return value of a lambda-defined function is the value of the expression itself; no return statement block is needed.
>5. The main use case of lambda expressions is to assign them to variables or pass them as arguments into other functions.

<!-- bilingual -->

```python
# def函数
def func(x):
    return x**2
# lambda函数
f = lambda x: x**2
print(type(f))
```

## 高阶函数 / Higher-Order Functions

---

接受一个或多个函数作为输入的一个函数称为高阶函数。

A function that takes one or more functions as input is called a higher-order function.

### map函数 / The map function

接收一个函数 f 和一个 list，并通过把函数 f 依次作用在 list 的每个元素上，并返回一个迭代器。

Takes a function f and a list, applies the function f in turn to each element of the list, and returns an iterator.

<!-- bilingual -->

```python
f = lambda x, y: (x ** y, x+y)
list_result = map(f, [2, 4, 6], [3, 2, 1])
print(list(list_result)) # 第一次遍历返回所有结果
print(list(list_result)) # 第二次遍历为空
```

### filter函数 / The filter function

接收一个函数 f 和一个 list，这个函数 f 的作用是对每个元素进行判断，返回 True 或 False，filter() 根据判断结果自动过滤掉不符合条件的元素，并返回一个迭代器。

Takes a function f and a list. The function f is used to test each element and returns True or False. filter() automatically filters out elements that do not satisfy the condition based on this result and returns an iterator.

<!-- bilingual -->

```python
def is_not_empty(s):
    return s and len(s.strip()) > 0

print(list(filter(is_not_empty, ['test', None, '', 'str', ' ', 'END'])))
```
