Python is a **programming language**, similar to C, Fortran, BASIC, PHP, and others. Some features of Python:

* 一种*解释性*（不是编译）语言。与C或者Fortran等不同，Python代码在执行前不会编译。另外，Python可以**交互**使用：有许多的Python解释器，命令和脚本可以在其中执行。

* 非常简单的接口与其他语言交互，特别是C和C++
* 稍后会介绍一些语言的其他特性。例如Python是面向对象的语言，包含动态类型（一个变量可以在程序过程中，可以包含不同的对象类型）。

* An *interpreted* (not compiled) language. Unlike C or Fortran, Python code is not compiled before being executed. In addition, Python can be used **interactively**: there are many Python interpreters in which commands and scripts can be executed.
* Free software released under an **open-source** license: Python can be used and distributed for free, even for commercial purposes.
* **Cross-platform**: Python can be used on all major operating systems — Windows, Linux/Unix, macOS X, and possibly even your phone's operating system, and so on.
* A very readable language, with clean, non-verbose syntax.
* Has a large number of high-quality packages that can be applied to a wide range of applications, from web frameworks to scientific computing.
* Very simple interfaces for interoperating with other languages, especially C and C++.
* A few other features of the language are introduced later. For example, Python is an object-oriented language and supports dynamic typing (a variable can contain different kinds of objects as a program runs).

### 值传递 / Value Passing

可以在一个函数内部改变变量的值吗？大多数语言（C、Java...）区分了“值传递“和”引用传递“。在Python中，没有严格的这种区分，并且视你的变量是否会修改而有一些不同。幸运的是，这些情况存在明确的规则。

Can you change the value of a variable inside a function? Most languages (C, Java, ...) distinguish between "pass by value" and "pass by reference." In Python, there is no strict distinction, and the behavior depends on whether your variable is modified. Fortunately, there are clear rules for these situations.

函数的参数是对象的引用，传递的是值。当你向一个函数传递了一个变量，Python传递的是对象的引用，这个对象引用的变量（值）。而不是变量本身。

A function's parameters are references to objects, passed by value. When you pass a variable to a function, Python passes the reference to the object — the object referenced by the variable (the value) — rather than the variable itself.

如果**值**传递给函数的值是不可变的，那么这个函数并不会改变调用者的变量。如果**值**是可变的，那么函数将可能在原地修改调用者的变量。

If the **value** passed to the function is immutable, then the function cannot change the caller's variable. If the **value** is mutable, then the function may modify the caller's variable in place.

<!-- bilingual -->

````python
def try_to_modify(x, y, z):
    x = 23
    y.append(42)
    z = [99] # 新引用
    print(x)
    print(y)
    print(z)

a = 77    # 不可变变量
b = [99]  # 可变变量
c = [28]
try_to_modify(a, b, c)
````

````python
print(a)
````

````python
print(b)
````

````python
print(c)
````

函数有名为*local namespace*的本地变量表。

A function has a local variable table called the *local namespace*.

变量X只存在于函数try_to_modify内部。

The variable X exists only inside the function try_to_modify.
