Jupyter Notebook (formerly IPython Notebook) is a web-based interactive computing environment used to create Jupyter Notebook documents. A Jupyter Notebook document is a JSON document containing an ordered list of input/output cells. These cells can contain code, text (using Markdown), mathematics, charts, and rich media, and the document usually has the extension ".ipynb".

### 执行脚本 / Executing Scripts

<!-- bilingual -->

```python
%%writefile /tmp/fib.py

def fib(N): 
    """ 
    Return a list of the first N Fibonacci numbers.
    """ 
    f0, f1 = 0, 1
    f = [1] * N
    for n in range(1, N):
        f[n] = f0 + f1
        f0, f1 = f1, f[n]

    return f

print(fib(10))
```

```python
!python /tmp/fib.py
```

```python
%run /tmp/fib.py
```

```python
!rm /tmp/fib.py
```

```python
fib(6)
```

```python
import numpy as np

def random_walker_max_distance(M, N):
    """
    Simulate N random walkers taking M steps, and return the largest distance
    from the starting point achieved by any of the random walkers.
    """
    trajectories = [np.random.randn(M).cumsum() for _ in range(N)]
    return np.max(np.abs(trajectories))
```

```python
%prun random_walker_max_distance(400, 10000)
```

```python
import numpy as np
from ipywidgets import interactive
import matplotlib.pyplot as plt

def f(a, b):
    plt.figure()
    x = np.linspace(-10,10,100)
    plt.plot(x, a * np.sin(b*x))
    plt.ylim(-2, 2)
    plt.show()

interact_plot = interactive(f, a=(1.0, 2.0), b=(1.0, 5.0))
interact_plot
```

```python
from ipywidgets import FloatSlider, FloatText, jslink

slider = FloatSlider(value = 7.5, min = 5.0, max = 10.0, step = 0.1, description = 'Input')
```

```python
text = FloatText(description = 'Value')
jslink((slider, 'value'), (text, 'value'))
text
```
